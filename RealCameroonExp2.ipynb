```python
# ============================================================
# EXPERIMENT 2
#
# Recursive sensitivities, uncertainty geometry,
# aleatory-versus-epistemic attribution,
# and surveillance-information priorities
#
# FIXED INPUTS FROM FINAL EXPERIMENT 1
#
# No data loading
# No parameter refitting
# No likelihood optimization
# No numerical Hessian estimation
#
# One autonomous Google Colab cell
# ============================================================

from __future__ import annotations

import time
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from IPython.display import display


TOTAL_START = time.perf_counter()


# ============================================================
# 1. FIXED RESULTS FROM EXPERIMENT 1
# ============================================================

# Selected discrete configuration

N = 175
S0 = 114
I0 = 1
R0_INITIAL = 60

assert N == S0 + I0 + R0_INITIAL


# Exact conditional estimates from Experiment 1

B_HAT = 0.7650254997116541
GAMMA_HAT = 0.23768509187018172
PHI_HAT = 2.4906108661339097

EXACT_PENALIZED_NLL = 45.351238703709555


# Fixed covariance matrix from Experiment 1
#
# Parameter ordering
#
# eta = (log b, log gamma, log phi)
#
# The matrix below is NOT re-estimated in Experiment 2.

SIGMA_ETA = np.array(
    [
        [
            0.01078625,
            0.01032878,
            -0.01802093,
        ],
        [
            0.01032878,
            0.01670211,
            -0.03299136,
        ],
        [
            -0.01802093,
            -0.03299136,
            0.55918411,
        ],
    ],
    dtype=float,
)


# The Hessian is obtained algebraically from the fixed
# Experiment 1 covariance matrix.
#
# This is not a new numerical Hessian calculation.

HESSIAN_ETA = np.linalg.inv(
    SIGMA_ETA
)


# Only b and gamma affect the exact stochastic descriptors.
# phi belongs to the observation model.

SIGMA_BG = SIGMA_ETA[
    np.ix_(
        [0, 1],
        [0, 1],
    )
]


ETA_BG_HAT = np.log(
    [
        B_HAT,
        GAMMA_HAT,
    ]
)


# Experiment 1 max-t critical value.
#
# This is used only as a fixed reference when translating
# hypothetical covariance reductions into risk-band widths.

EXP1_MAX_T_CRITICAL = 3.8270510451310362


# Experiment 1 reference quantities used as numerical checks

EXP1_FINAL_SIZE_MEAN = 47.624525
EXP1_EXTINCTION_MEAN = 23.658440

EXP1_TAIL_RISKS = np.array(
    [
        0.502185,
        0.492970,
        0.454210,
        0.318310,
        0.045407,
    ],
    dtype=float,
)


# Thresholds are exactly the same fractions of S(0)
# used in Experiment 1.

THRESHOLD_FRACTIONS = np.array(
    [
        0.50,
        0.60,
        0.70,
        0.80,
        0.90,
    ],
    dtype=float,
)

THRESHOLDS = np.round(
    THRESHOLD_FRACTIONS
    * S0
).astype(int)

assert np.array_equal(
    THRESHOLDS,
    np.array(
        [57, 68, 80, 91, 103]
    ),
)


# Numerical validation settings

FINITE_DIFFERENCE_STEP = 1e-5


# Output directory

OUTPUT_DIR = Path(
    "/content/experiment2_fixed_from_experiment1"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


print(
    "Fixed Experiment 1 configuration"
)

print(
    {
        "N_eff": N,
        "S0": S0,
        "I0": I0,
        "R0": R0_INITIAL,
        "b_hat": B_HAT,
        "gamma_hat": GAMMA_HAT,
        "phi_hat": PHI_HAT,
        "exact_penalized_nll":
            EXACT_PENALIZED_NLL,
    }
)

print()

print(
    "Fixed covariance matrix"
)

print(
    SIGMA_ETA
)

print()

print(
    "Corresponding Hessian"
)

print(
    HESSIAN_ETA
)

print()

print(
    "Hessian eigenvalues"
)

print(
    np.linalg.eigvalsh(
        HESSIAN_ETA
    )
)


# ============================================================
# 2. EXACT FINAL-SIZE PMF AND RECURSIVE SENSITIVITIES
#
# Derivatives are with respect to
#
# eta_b     = log b
# eta_gamma = log gamma
# ============================================================

def final_size_pmf_gradient(
    b: float,
    gamma: float,
) -> tuple[
    np.ndarray,
    np.ndarray,
]:
    """
    Exact final-size PMF and recursive derivatives with
    respect to log b and log gamma.

    pmf[n] = P(N^I = n)

    gradient[k,n] is the derivative of pmf[n] with respect
    to log b for k=0 and log gamma for k=1.
    """

    probability = np.zeros(
        I0 + 1,
        dtype=float,
    )

    probability[I0] = 1.0

    gradient = np.zeros(
        (
            2,
            I0 + 1,
        ),
        dtype=float,
    )

    pmf = np.zeros(
        S0 + 1,
        dtype=float,
    )

    pmf_gradient = np.zeros(
        (
            2,
            S0 + 1,
        ),
        dtype=float,
    )


    for susceptible in range(
        S0,
        -1,
        -1,
    ):

        maximum_infectious = (
            I0
            + S0
            - susceptible
        )

        required_length = (
            maximum_infectious
            + 1
        )

        if len(
            probability
        ) < required_length:

            padding = (
                required_length
                - len(
                    probability
                )
            )

            probability = np.pad(
                probability,
                (
                    0,
                    padding,
                ),
            )

            gradient = np.pad(
                gradient,
                (
                    (0, 0),
                    (0, padding),
                ),
            )


        if susceptible > 0:

            infection_component = (
                b
                * susceptible
                / N
            )

            denominator = (
                infection_component
                + gamma
            )

            infection_probability = (
                infection_component
                / denominator
            )

            recovery_probability = (
                gamma
                / denominator
            )


            # Derivatives of the event probabilities
            # with respect to log b and log gamma

            derivative_infection = np.array(
                [
                    infection_probability
                    * recovery_probability,

                    -infection_probability
                    * recovery_probability,
                ],
                dtype=float,
            )

            derivative_recovery = (
                -derivative_infection
            )


            next_probability = np.zeros(
                maximum_infectious
                + 2,
                dtype=float,
            )

            next_gradient = np.zeros(
                (
                    2,
                    maximum_infectious
                    + 2,
                ),
                dtype=float,
            )

        else:

            infection_probability = 0.0
            recovery_probability = 1.0

            derivative_infection = np.zeros(
                2,
                dtype=float,
            )

            derivative_recovery = np.zeros(
                2,
                dtype=float,
            )

            next_probability = None
            next_gradient = None


        # Descending order allows repeated recoveries
        # within a fixed susceptible level.

        for infectious in range(
            maximum_infectious,
            0,
            -1,
        ):

            mass = probability[
                infectious
            ]

            mass_gradient = gradient[
                :,
                infectious,
            ].copy()

            if (
                abs(mass) < 1e-300
                and np.all(
                    np.abs(
                        mass_gradient
                    )
                    < 1e-300
                )
            ):
                continue


            # Recovery transition

            probability[
                infectious - 1
            ] += (
                mass
                * recovery_probability
            )

            gradient[
                :,
                infectious - 1,
            ] += (
                mass_gradient
                * recovery_probability
                + mass
                * derivative_recovery
            )


            # Infection transition

            if susceptible > 0:

                next_probability[
                    infectious + 1
                ] += (
                    mass
                    * infection_probability
                )

                next_gradient[
                    :,
                    infectious + 1,
                ] += (
                    mass_gradient
                    * infection_probability
                    + mass
                    * derivative_infection
                )


        number_of_new_infections = (
            S0
            - susceptible
        )

        pmf[
            number_of_new_infections
        ] = probability[0]

        pmf_gradient[
            :,
            number_of_new_infections,
        ] = gradient[
            :,
            0,
        ]


        if susceptible > 0:

            probability = (
                next_probability
            )

            gradient = (
                next_gradient
            )


    # Numerical normalization

    normalizer = float(
        pmf.sum()
    )

    normalizer_gradient = (
        pmf_gradient.sum(
            axis=1
        )
    )

    normalized_pmf = (
        pmf
        / normalizer
    )

    normalized_gradient = (
        pmf_gradient
        * normalizer
        - pmf[
            None,
            :
        ]
        * normalizer_gradient[
            :,
            None,
        ]
    ) / (
        normalizer**2
    )


    return (
        normalized_pmf,
        normalized_gradient,
    )


# ============================================================
# 3. STRICT EXCEEDANCE PROBABILITIES
#
# rho(c) = P(N^I > c)
# ============================================================

def strict_tail_and_gradient(
    pmf: np.ndarray,
    pmf_gradient: np.ndarray,
) -> tuple[
    np.ndarray,
    np.ndarray,
]:

    nonstrict_tail = np.cumsum(
        pmf[::-1]
    )[::-1]

    nonstrict_gradient = np.cumsum(
        pmf_gradient[
            :,
            ::-1,
        ],
        axis=1,
    )[
        :,
        ::-1,
    ]


    strict_tail = (
        nonstrict_tail
        - pmf
    )

    strict_gradient = (
        nonstrict_gradient
        - pmf_gradient
    )


    return (
        strict_tail,
        strict_gradient,
    )


# ============================================================
# 4. EXTINCTION-TIME MEAN AND RECURSIVE SENSITIVITY
# ============================================================

def extinction_mean_gradient(
    b: float,
    gamma: float,
) -> tuple[
    float,
    np.ndarray,
]:
    """
    Exact expected extinction time and its recursive
    derivatives with respect to log b and log gamma.
    """

    previous_mean = None
    previous_gradient = None


    for susceptible in range(
        0,
        S0 + 1,
    ):

        maximum_infectious = (
            I0
            + S0
            - susceptible
        )

        mean = np.zeros(
            maximum_infectious
            + 1,
            dtype=float,
        )

        gradient = np.zeros(
            (
                2,
                maximum_infectious
                + 1,
            ),
            dtype=float,
        )


        infection_component = (
            b
            * susceptible
            / N
        )

        denominator = (
            infection_component
            + gamma
        )


        if susceptible > 0:

            infection_probability = (
                infection_component
                / denominator
            )

        else:

            infection_probability = 0.0


        recovery_probability = (
            1.0
            - infection_probability
        )


        derivative_infection = np.array(
            [
                infection_probability
                * recovery_probability,

                -infection_probability
                * recovery_probability,
            ],
            dtype=float,
        )


        for infectious in range(
            1,
            maximum_infectious
            + 1,
        ):

            total_rate = (
                infectious
                * denominator
            )

            holding_mean = (
                1.0
                / total_rate
            )


            # Derivative of the holding-time mean
            # on the log-parameter scale

            holding_gradient = (
                -holding_mean
                * np.array(
                    [
                        infection_probability,
                        recovery_probability,
                    ],
                    dtype=float,
                )
            )


            recovery_mean = mean[
                infectious - 1
            ]

            recovery_gradient = gradient[
                :,
                infectious - 1,
            ]


            if susceptible > 0:

                infection_mean = (
                    previous_mean[
                        infectious + 1
                    ]
                )

                infection_gradient = (
                    previous_gradient[
                        :,
                        infectious + 1,
                    ]
                )

            else:

                infection_mean = 0.0

                infection_gradient = (
                    np.zeros(
                        2,
                        dtype=float,
                    )
                )


            next_mean = (
                infection_probability
                * infection_mean
                + recovery_probability
                * recovery_mean
            )


            next_gradient = (
                derivative_infection
                * (
                    infection_mean
                    - recovery_mean
                )
                + infection_probability
                * infection_gradient
                + recovery_probability
                * recovery_gradient
            )


            mean[
                infectious
            ] = (
                holding_mean
                + next_mean
            )


            gradient[
                :,
                infectious,
            ] = (
                holding_gradient
                + next_gradient
            )


        previous_mean = mean
        previous_gradient = gradient


    return (
        float(
            mean[I0]
        ),
        gradient[
            :,
            I0,
        ].copy(),
    )


# ============================================================
# 5. EXACT EXTINCTION-TIME MOMENTS
#
# Required for aleatory uncertainty of extinction time.
# ============================================================

def extinction_time_moments(
    b: float,
    gamma: float,
) -> dict:

    previous_mean = None
    previous_second = None


    for susceptible in range(
        0,
        S0 + 1,
    ):

        maximum_infectious = (
            I0
            + S0
            - susceptible
        )

        mean = np.zeros(
            maximum_infectious
            + 1,
            dtype=float,
        )

        second = np.zeros(
            maximum_infectious
            + 1,
            dtype=float,
        )


        infection_component = (
            b
            * susceptible
            / N
        )

        denominator = (
            infection_component
            + gamma
        )


        infection_probability = (
            infection_component
            / denominator
            if susceptible > 0
            else 0.0
        )

        recovery_probability = (
            1.0
            - infection_probability
        )


        for infectious in range(
            1,
            maximum_infectious
            + 1,
        ):

            total_rate = (
                infectious
                * denominator
            )

            holding_mean = (
                1.0
                / total_rate
            )


            recovery_mean = mean[
                infectious - 1
            ]

            recovery_second = second[
                infectious - 1
            ]


            if susceptible > 0:

                infection_mean = (
                    previous_mean[
                        infectious + 1
                    ]
                )

                infection_second = (
                    previous_second[
                        infectious + 1
                    ]
                )

            else:

                infection_mean = 0.0
                infection_second = 0.0


            next_mean = (
                infection_probability
                * infection_mean
                + recovery_probability
                * recovery_mean
            )

            next_second = (
                infection_probability
                * infection_second
                + recovery_probability
                * recovery_second
            )


            mean[
                infectious
            ] = (
                holding_mean
                + next_mean
            )


            second[
                infectious
            ] = (
                2.0
                * holding_mean**2

                + 2.0
                * holding_mean
                * next_mean

                + next_second
            )


        previous_mean = mean
        previous_second = second


    mean_value = float(
        mean[I0]
    )

    second_value = float(
        second[I0]
    )

    variance_value = max(
        second_value
        - mean_value**2,
        0.0,
    )


    return {
        "mean":
            mean_value,

        "variance":
            variance_value,

        "sd":
            np.sqrt(
                variance_value
            ),
    }


# ============================================================
# 6. ALL DESCRIPTORS AND RECURSIVE GRADIENTS
# ============================================================

def descriptors_and_gradients(
    eta_bg: np.ndarray,
) -> tuple[
    np.ndarray,
    np.ndarray,
    dict,
]:

    b = float(
        np.exp(
            eta_bg[0]
        )
    )

    gamma = float(
        np.exp(
            eta_bg[1]
        )
    )


    (
        pmf,
        pmf_gradient,
    ) = final_size_pmf_gradient(
        b=b,
        gamma=gamma,
    )


    (
        strict_tail,
        strict_tail_gradient,
    ) = strict_tail_and_gradient(
        pmf=pmf,
        pmf_gradient=
            pmf_gradient,
    )


    final_size_values = np.arange(
        len(
            pmf
        ),
        dtype=float,
    )


    final_size_mean = float(
        np.sum(
            final_size_values
            * pmf
        )
    )


    final_size_mean_gradient = np.sum(
        final_size_values[
            None,
            :
        ]
        * pmf_gradient,
        axis=1,
    )


    final_size_second = float(
        np.sum(
            final_size_values**2
            * pmf
        )
    )

    final_size_variance = max(
        final_size_second
        - final_size_mean**2,
        0.0,
    )


    (
        extinction_mean,
        extinction_gradient,
    ) = extinction_mean_gradient(
        b=b,
        gamma=gamma,
    )


    descriptors = np.concatenate(
        [
            strict_tail[
                THRESHOLDS
            ],

            np.array(
                [
                    final_size_mean,
                    extinction_mean,
                ],
                dtype=float,
            ),
        ]
    )


    gradient_matrix = np.vstack(
        [
            strict_tail_gradient[
                :,
                THRESHOLDS,
            ].T,

            final_size_mean_gradient,

            extinction_gradient,
        ]
    )


    additional = {

        "pmf":
            pmf,

        "pmf_gradient":
            pmf_gradient,

        "strict_tail":
            strict_tail,

        "strict_tail_gradient":
            strict_tail_gradient,

        "final_size_mean":
            final_size_mean,

        "final_size_variance":
            final_size_variance,

        "final_size_sd":
            np.sqrt(
                final_size_variance
            ),

        "extinction_mean":
            extinction_mean,
    }


    return (
        descriptors,
        gradient_matrix,
        additional,
    )


COMPUTATION_START = time.perf_counter()


(
    DESCRIPTORS,
    RECURSIVE_GRADIENT,
    ADDITIONAL,
) = descriptors_and_gradients(
    ETA_BG_HAT
)


EXTINCTION_MOMENTS = (
    extinction_time_moments(
        b=B_HAT,
        gamma=GAMMA_HAT,
    )
)


TAIL_RISKS = DESCRIPTORS[
    : len(
        THRESHOLDS
    )
]

FINAL_SIZE_MEAN = DESCRIPTORS[
    -2
]

EXTINCTION_MEAN = DESCRIPTORS[
    -1
]


DESCRIPTOR_NAMES = [

    *[
        f"P(NI > {threshold})"
        for threshold
        in THRESHOLDS
    ],

    "Expected final size",

    "Expected extinction time",
]


# ============================================================
# 7. INTERNAL CONSISTENCY CHECK AGAINST EXPERIMENT 1
# ============================================================

print()
print(
    "Experiment 1 consistency checks"
)


print(
    {
        "max_tail_risk_difference":
            float(
                np.max(
                    np.abs(
                        TAIL_RISKS
                        - EXP1_TAIL_RISKS
                    )
                )
            ),

        "final_size_difference":
            float(
                FINAL_SIZE_MEAN
                - EXP1_FINAL_SIZE_MEAN
            ),

        "extinction_mean_difference":
            float(
                EXTINCTION_MEAN
                - EXP1_EXTINCTION_MEAN
            ),
    }
)


if np.max(
    np.abs(
        TAIL_RISKS
        - EXP1_TAIL_RISKS
    )
) > 5e-4:

    raise RuntimeError(
        "Experiment 2 does not reproduce the "
        "Experiment 1 tail risks."
    )


if abs(
    FINAL_SIZE_MEAN
    - EXP1_FINAL_SIZE_MEAN
) > 5e-4:

    raise RuntimeError(
        "Experiment 2 does not reproduce the "
        "Experiment 1 expected final size."
    )


if abs(
    EXTINCTION_MEAN
    - EXP1_EXTINCTION_MEAN
) > 5e-4:

    raise RuntimeError(
        "Experiment 2 does not reproduce the "
        "Experiment 1 extinction mean."
    )


# ============================================================
# 8. VALIDATE RECURSIVE SENSITIVITIES
#    AGAINST CENTRAL FINITE DIFFERENCES
# ============================================================

FINITE_DIFFERENCE_GRADIENT = np.zeros_like(
    RECURSIVE_GRADIENT
)


for parameter_index in range(
    2
):

    direction = np.zeros(
        2,
        dtype=float,
    )

    direction[
        parameter_index
    ] = (
        FINITE_DIFFERENCE_STEP
    )


    descriptors_plus, _, _ = (
        descriptors_and_gradients(
            ETA_BG_HAT
            + direction
        )
    )

    descriptors_minus, _, _ = (
        descriptors_and_gradients(
            ETA_BG_HAT
            - direction
        )
    )


    FINITE_DIFFERENCE_GRADIENT[
        :,
        parameter_index,
    ] = (
        descriptors_plus
        - descriptors_minus
    ) / (
        2.0
        * FINITE_DIFFERENCE_STEP
    )


GRADIENT_ERROR = np.abs(
    RECURSIVE_GRADIENT
    - FINITE_DIFFERENCE_GRADIENT
)


RELATIVE_GRADIENT_ERROR = (
    GRADIENT_ERROR
    / np.maximum(
        np.abs(
            FINITE_DIFFERENCE_GRADIENT
        ),
        1e-12,
    )
)


GRADIENT_VALIDATION = pd.DataFrame(
    {
        "descriptor":
            DESCRIPTOR_NAMES,

        "recursive_log_b":
            RECURSIVE_GRADIENT[
                :,
                0,
            ],

        "finite_difference_log_b":
            FINITE_DIFFERENCE_GRADIENT[
                :,
                0,
            ],

        "absolute_error_log_b":
            GRADIENT_ERROR[
                :,
                0,
            ],

        "relative_error_log_b":
            RELATIVE_GRADIENT_ERROR[
                :,
                0,
            ],

        "recursive_log_gamma":
            RECURSIVE_GRADIENT[
                :,
                1,
            ],

        "finite_difference_log_gamma":
            FINITE_DIFFERENCE_GRADIENT[
                :,
                1,
            ],

        "absolute_error_log_gamma":
            GRADIENT_ERROR[
                :,
                1,
            ],

        "relative_error_log_gamma":
            RELATIVE_GRADIENT_ERROR[
                :,
                1,
            ],
    }
)


print()
print(
    "Recursive sensitivity validation"
)

display(
    GRADIENT_VALIDATION.round(
        10
    )
)


# ============================================================
# 9. DELTA-METHOD DESCRIPTOR COVARIANCE
# ============================================================

DESCRIPTOR_COVARIANCE = (
    RECURSIVE_GRADIENT
    @ SIGMA_BG
    @ RECURSIVE_GRADIENT.T
)


DESCRIPTOR_STANDARD_ERRORS = np.sqrt(
    np.maximum(
        np.diag(
            DESCRIPTOR_COVARIANCE
        ),
        0.0,
    )
)


TAIL_STANDARD_ERRORS = (
    DESCRIPTOR_STANDARD_ERRORS[
        : len(
            THRESHOLDS
        )
    ]
)


# ============================================================
# 10. UNCERTAINTY GEOMETRY
#
# The largest eigen-direction of Sigma_BG represents the
# dominant local parameter-uncertainty direction.
#
# We quantify how much each descriptor gradient projects
# onto that direction.
# ============================================================

BG_EIGENVALUES, BG_EIGENVECTORS = (
    np.linalg.eigh(
        SIGMA_BG
    )
)


DOMINANT_INDEX = int(
    np.argmax(
        BG_EIGENVALUES
    )
)


DOMINANT_EIGENVALUE = float(
    BG_EIGENVALUES[
        DOMINANT_INDEX
    ]
)


DOMINANT_DIRECTION = (
    BG_EIGENVECTORS[
        :,
        DOMINANT_INDEX
    ]
)


# Fix sign only for reproducibility of printing

if DOMINANT_DIRECTION[0] < 0:

    DOMINANT_DIRECTION = (
        -DOMINANT_DIRECTION
    )


# ============================================================
# 11. ALEATORY VERSUS EPISTEMIC ATTRIBUTION
# ============================================================

TAIL_ALEATORY_VARIANCE = (
    TAIL_RISKS
    * (
        1.0
        - TAIL_RISKS
    )
)


FINAL_SIZE_ALEATORY_VARIANCE = (
    ADDITIONAL[
        "final_size_variance"
    ]
)


EXTINCTION_ALEATORY_VARIANCE = (
    EXTINCTION_MOMENTS[
        "variance"
    ]
)


ALEATORY_VARIANCES = np.concatenate(
    [
        TAIL_ALEATORY_VARIANCE,

        np.array(
            [
                FINAL_SIZE_ALEATORY_VARIANCE,
                EXTINCTION_ALEATORY_VARIANCE,
            ]
        ),
    ]
)


EPISTEMIC_VARIANCES = (
    DESCRIPTOR_STANDARD_ERRORS**2
)


KAPPA = np.divide(
    EPISTEMIC_VARIANCES,
    ALEATORY_VARIANCES,
    out=np.full_like(
        EPISTEMIC_VARIANCES,
        np.nan,
    ),
    where=(
        ALEATORY_VARIANCES
        > 1e-12
    ),
)


ATTRIBUTION_ROWS = []


for descriptor_index, descriptor_name in enumerate(
    DESCRIPTOR_NAMES
):

    gradient = (
        RECURSIVE_GRADIENT[
            descriptor_index
        ]
    )


    epistemic_variance = float(
        EPISTEMIC_VARIANCES[
            descriptor_index
        ]
    )


    # Exact first-order variance decomposition
    #
    # V =
    # gb^2 Var(log b)
    # +
    # gg^2 Var(log gamma)
    # +
    # 2 gb gg Cov(log b,log gamma)

    transmission_diagonal = float(
        gradient[0]**2
        * SIGMA_BG[
            0,
            0,
        ]
    )

    recovery_diagonal = float(
        gradient[1]**2
        * SIGMA_BG[
            1,
            1,
        ]
    )

    covariance_component = float(
        2.0
        * gradient[0]
        * gradient[1]
        * SIGMA_BG[
            0,
            1,
        ]
    )


    gradient_norm = float(
        np.linalg.norm(
            gradient
        )
    )


    if gradient_norm > 1e-14:

        cosine_alignment = float(
            abs(
                gradient
                @ DOMINANT_DIRECTION
            )
            / gradient_norm
        )

    else:

        cosine_alignment = np.nan


    if epistemic_variance > 1e-14:

        dominant_direction_share = float(
            DOMINANT_EIGENVALUE
            * (
                gradient
                @ DOMINANT_DIRECTION
            )**2
            / epistemic_variance
        )

    else:

        dominant_direction_share = (
            np.nan
        )


    ATTRIBUTION_ROWS.append(
        {
            "descriptor":
                descriptor_name,

            "estimate":
                DESCRIPTORS[
                    descriptor_index
                ],

            "gradient_log_b":
                gradient[0],

            "gradient_log_gamma":
                gradient[1],

            "delta_standard_error":
                DESCRIPTOR_STANDARD_ERRORS[
                    descriptor_index
                ],

            "aleatory_variance":
                ALEATORY_VARIANCES[
                    descriptor_index
                ],

            "epistemic_variance":
                epistemic_variance,

            "kappa_epistemic_to_aleatory":
                KAPPA[
                    descriptor_index
                ],

            "transmission_diagonal_component":
                transmission_diagonal,

            "recovery_diagonal_component":
                recovery_diagonal,

            "cross_covariance_component":
                covariance_component,

            "gradient_alignment_with_dominant_direction":
                cosine_alignment,

            "dominant_eigendirection_variance_share":
                dominant_direction_share,
        }
    )


ATTRIBUTION_SUMMARY = pd.DataFrame(
    ATTRIBUTION_ROWS
)


TAIL_ATTRIBUTION = (
    ATTRIBUTION_SUMMARY.iloc[
        : len(
            THRESHOLDS
        )
    ]
    .copy()
)


TAIL_ATTRIBUTION.insert(
    1,
    "threshold",
    THRESHOLDS,
)


TAIL_ATTRIBUTION.insert(
    2,
    "fraction_of_S0",
    THRESHOLDS
    / S0,
)


print()
print(
    "Uncertainty attribution"
)

display(
    ATTRIBUTION_SUMMARY.round(
        6
    )
)


# ============================================================
# 12. CURRENT TAIL-RISK SUMMARY
#
# This is included only to verify agreement with Experiment 1.
# Experiment 2 focuses on sensitivities and attribution.
# ============================================================

CURRENT_TAIL_SUMMARY = pd.DataFrame(
    {
        "threshold":
            THRESHOLDS,

        "fraction_of_S0":
            THRESHOLDS
            / S0,

        "tail_risk":
            TAIL_RISKS,

        "delta_standard_error":
            TAIL_STANDARD_ERRORS,

        "aleatory_variance":
            TAIL_ALEATORY_VARIANCE,

        "epistemic_variance":
            EPISTEMIC_VARIANCES[
                : len(
                    THRESHOLDS
                )
            ],

        "kappa":
            KAPPA[
                : len(
                    THRESHOLDS
                )
            ],
    }
)


print()
print(
    "Tail-risk attribution summary"
)

display(
    CURRENT_TAIL_SUMMARY.round(
        6
    )
)


# ============================================================
# 13. SURVEILLANCE-INFORMATION SCENARIOS
#
# These scenarios do not modify b or gamma.
#
# They modify only their estimation uncertainty.
#
# D Sigma D preserves the baseline correlation structure while
# scaling the marginal standard errors.
# ============================================================

def scale_covariance(
    covariance: np.ndarray,
    b_se_multiplier: float = 1.0,
    gamma_se_multiplier: float = 1.0,
) -> np.ndarray:

    scaling = np.diag(
        [
            b_se_multiplier,
            gamma_se_multiplier,
        ]
    )

    return (
        scaling
        @ covariance
        @ scaling
    )


SURVEILLANCE_SCENARIOS = {

    "Current information":
        SIGMA_BG,

    "Transmission SE reduced 25%":
        scale_covariance(
            SIGMA_BG,
            b_se_multiplier=0.75,
        ),

    "Transmission SE reduced 50%":
        scale_covariance(
            SIGMA_BG,
            b_se_multiplier=0.50,
        ),

    "Recovery SE reduced 25%":
        scale_covariance(
            SIGMA_BG,
            gamma_se_multiplier=0.75,
        ),

    "Recovery SE reduced 50%":
        scale_covariance(
            SIGMA_BG,
            gamma_se_multiplier=0.50,
        ),

    "Both SEs reduced 25%":
        scale_covariance(
            SIGMA_BG,
            b_se_multiplier=0.75,
            gamma_se_multiplier=0.75,
        ),

    "Double effective information":
        SIGMA_BG
        / 2.0,
}


SURVEILLANCE_ROWS = []


for scenario_name, scenario_covariance in (
    SURVEILLANCE_SCENARIOS.items()
):

    scenario_variances = np.maximum(
        np.einsum(
            "ki,ij,kj->k",
            RECURSIVE_GRADIENT,
            scenario_covariance,
            RECURSIVE_GRADIENT,
        ),
        0.0,
    )


    scenario_standard_errors = np.sqrt(
        scenario_variances
    )


    scenario_tail_se = (
        scenario_standard_errors[
            : len(
                THRESHOLDS
            )
        ]
    )


    # For comparability with Experiment 1, the same
    # bootstrap max-t critical value is held fixed.
    #
    # This isolates the effect of improved parameter
    # information on the risk-band width.

    scenario_lower = np.clip(
        TAIL_RISKS
        - EXP1_MAX_T_CRITICAL
        * scenario_tail_se,
        0.0,
        1.0,
    )

    scenario_upper = np.clip(
        TAIL_RISKS
        + EXP1_MAX_T_CRITICAL
        * scenario_tail_se,
        0.0,
        1.0,
    )

    scenario_width = (
        scenario_upper
        - scenario_lower
    )


    SURVEILLANCE_ROWS.append(
        {
            "scenario":
                scenario_name,

            "mean_tail_standard_error":
                float(
                    np.mean(
                        scenario_tail_se
                    )
                ),

            "maximum_tail_standard_error":
                float(
                    np.max(
                        scenario_tail_se
                    )
                ),

            "mean_reference_max_t_band_width":
                float(
                    np.mean(
                        scenario_width
                    )
                ),

            "maximum_reference_max_t_band_width":
                float(
                    np.max(
                        scenario_width
                    )
                ),

            "expected_final_size_standard_error":
                float(
                    scenario_standard_errors[
                        -2
                    ]
                ),

            "expected_extinction_time_standard_error":
                float(
                    scenario_standard_errors[
                        -1
                    ]
                ),
        }
    )


SURVEILLANCE_SUMMARY = pd.DataFrame(
    SURVEILLANCE_ROWS
)


CURRENT_INFORMATION = (
    SURVEILLANCE_SUMMARY.loc[
        SURVEILLANCE_SUMMARY[
            "scenario"
        ]
        == "Current information"
    ]
    .iloc[0]
)


SURVEILLANCE_SUMMARY[
    "tail_se_reduction_percent"
] = (
    100.0
    * (
        CURRENT_INFORMATION[
            "mean_tail_standard_error"
        ]
        - SURVEILLANCE_SUMMARY[
            "mean_tail_standard_error"
        ]
    )
    / CURRENT_INFORMATION[
        "mean_tail_standard_error"
    ]
)


SURVEILLANCE_SUMMARY[
    "tail_band_reduction_percent"
] = (
    100.0
    * (
        CURRENT_INFORMATION[
            "mean_reference_max_t_band_width"
        ]
        - SURVEILLANCE_SUMMARY[
            "mean_reference_max_t_band_width"
        ]
    )
    / CURRENT_INFORMATION[
        "mean_reference_max_t_band_width"
    ]
)


SURVEILLANCE_SUMMARY[
    "final_size_se_reduction_percent"
] = (
    100.0
    * (
        CURRENT_INFORMATION[
            "expected_final_size_standard_error"
        ]
        - SURVEILLANCE_SUMMARY[
            "expected_final_size_standard_error"
        ]
    )
    / CURRENT_INFORMATION[
        "expected_final_size_standard_error"
    ]
)


SURVEILLANCE_SUMMARY[
    "extinction_se_reduction_percent"
] = (
    100.0
    * (
        CURRENT_INFORMATION[
            "expected_extinction_time_standard_error"
        ]
        - SURVEILLANCE_SUMMARY[
            "expected_extinction_time_standard_error"
        ]
    )
    / CURRENT_INFORMATION[
        "expected_extinction_time_standard_error"
    ]
)


print()
print(
    "Surveillance-information scenarios"
)

display(
    SURVEILLANCE_SUMMARY.round(
        4
    )
)


# ============================================================
# 14. SUMMARY OF FINAL SIZE AND EXTINCTION DESCRIPTORS
# ============================================================

DESCRIPTOR_SUMMARY = pd.DataFrame(
    {
        "quantity": [

            "Expected final size",

            "Final-size variance",

            "Final-size standard deviation",

            "Expected extinction time",

            "Extinction-time variance",

            "Extinction-time standard deviation",

            "Epistemic SE of expected final size",

            "Epistemic SE of expected extinction time",
        ],

        "estimate": [

            FINAL_SIZE_MEAN,

            ADDITIONAL[
                "final_size_variance"
            ],

            ADDITIONAL[
                "final_size_sd"
            ],

            EXTINCTION_MOMENTS[
                "mean"
            ],

            EXTINCTION_MOMENTS[
                "variance"
            ],

            EXTINCTION_MOMENTS[
                "sd"
            ],

            DESCRIPTOR_STANDARD_ERRORS[
                -2
            ],

            DESCRIPTOR_STANDARD_ERRORS[
                -1
            ],
        ],
    }
)


print()
print(
    "Descriptor summary"
)

display(
    DESCRIPTOR_SUMMARY.round(
        6
    )
)


# ============================================================
# 15. FIGURE
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(
        13,
        9,
    ),
)


# ------------------------------------------------------------
# Panel A
# Recursive sensitivities
# ------------------------------------------------------------

axes[
    0,
    0,
].plot(
    THRESHOLDS,
    RECURSIVE_GRADIENT[
        : len(
            THRESHOLDS
        ),
        0,
    ],
    marker="o",
    linewidth=2,
    label=r"$\partial\rho(c)/\partial\log b$",
)


axes[
    0,
    0,
].plot(
    THRESHOLDS,
    RECURSIVE_GRADIENT[
        : len(
            THRESHOLDS
        ),
        1,
    ],
    marker="o",
    linewidth=2,
    label=r"$\partial\rho(c)/\partial\log\gamma$",
)


axes[
    0,
    0,
].axhline(
    0.0,
    color="black",
    linewidth=0.8,
)


axes[
    0,
    0,
].set_xlabel(
    "Final-size threshold"
)

axes[
    0,
    0,
].set_ylabel(
    "Recursive sensitivity"
)

axes[
    0,
    0,
].legend(
    frameon=False
)

axes[
    0,
    0,
].grid(
    alpha=0.20
)


# ------------------------------------------------------------
# Panel B
# Validation of recursive gradients
# ------------------------------------------------------------

axes[
    0,
    1,
].semilogy(
    THRESHOLDS,
    GRADIENT_ERROR[
        : len(
            THRESHOLDS
        ),
        0,
    ]
    + 1e-18,
    marker="o",
    linewidth=2,
    label=r"$\log b$",
)


axes[
    0,
    1,
].semilogy(
    THRESHOLDS,
    GRADIENT_ERROR[
        : len(
            THRESHOLDS
        ),
        1,
    ]
    + 1e-18,
    marker="o",
    linewidth=2,
    label=r"$\log\gamma$",
)


axes[
    0,
    1,
].set_xlabel(
    "Final-size threshold"
)

axes[
    0,
    1,
].set_ylabel(
    "Absolute recursive versus finite-difference error"
)

axes[
    0,
    1,
].legend(
    frameon=False
)

axes[
    0,
    1,
].grid(
    alpha=0.20
)


# ------------------------------------------------------------
# Panel C
# Epistemic-to-aleatory ratio
# ------------------------------------------------------------

axes[
    1,
    0,
].bar(
    THRESHOLDS,
    KAPPA[
        : len(
            THRESHOLDS
        )
    ],
    width=7,
    color="tab:purple",
)


axes[
    1,
    0,
].set_xlabel(
    "Final-size threshold"
)

axes[
    1,
    0,
].set_ylabel(
    r"$\widehat{\kappa}(c)$"
)

axes[
    1,
    0,
].grid(
    axis="y",
    alpha=0.20
)


# ------------------------------------------------------------
# Panel D
# Information-priority scenarios
# ------------------------------------------------------------

PLOT_SURVEILLANCE = (
    SURVEILLANCE_SUMMARY.loc[
        SURVEILLANCE_SUMMARY[
            "scenario"
        ]
        != "Current information"
    ]
    .copy()
)


PLOT_SURVEILLANCE = (
    PLOT_SURVEILLANCE
    .sort_values(
        "tail_se_reduction_percent"
    )
)


axes[
    1,
    1,
].barh(
    PLOT_SURVEILLANCE[
        "scenario"
    ],
    PLOT_SURVEILLANCE[
        "tail_se_reduction_percent"
    ],
    color="tab:blue",
)


axes[
    1,
    1,
].set_xlabel(
    "Reduction in mean tail-risk epistemic SE (%)"
)

axes[
    1,
    1,
].grid(
    axis="x",
    alpha=0.20
)


fig.tight_layout()


fig.savefig(
    OUTPUT_DIR
    / "experiment2_uncertainty_geometry.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 16. SECOND FIGURE
#     PARAMETER-UNCERTAINTY GEOMETRY
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        12,
        4.8,
    ),
)


axes[0].plot(
    THRESHOLDS,
    TAIL_ATTRIBUTION[
        "gradient_alignment_with_dominant_direction"
    ],
    marker="o",
    linewidth=2,
    color="tab:green",
)


axes[0].set_ylim(
    0.0,
    1.05,
)

axes[0].set_xlabel(
    "Final-size threshold"
)

axes[0].set_ylabel(
    "Gradient alignment with dominant parameter direction"
)

axes[0].grid(
    alpha=0.20
)


axes[1].plot(
    THRESHOLDS,
    TAIL_ATTRIBUTION[
        "dominant_eigendirection_variance_share"
    ],
    marker="o",
    linewidth=2,
    color="tab:red",
)


axes[1].set_ylim(
    0.0,
    1.05,
)

axes[1].set_xlabel(
    "Final-size threshold"
)

axes[1].set_ylabel(
    "Share of epistemic variance from dominant direction"
)

axes[1].grid(
    alpha=0.20
)


fig.tight_layout()


fig.savefig(
    OUTPUT_DIR
    / "experiment2_parameter_uncertainty_geometry.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 17. SAVE ALL RESULTS
# ============================================================

FIXED_INPUTS = pd.DataFrame(
    {
        "quantity": [
            "N_eff",
            "S0",
            "I0",
            "R0",
            "b_hat",
            "gamma_hat",
            "phi_hat",
            "exact_penalized_nll",
            "Experiment1_max_t_critical",
        ],

        "value": [
            N,
            S0,
            I0,
            R0_INITIAL,
            B_HAT,
            GAMMA_HAT,
            PHI_HAT,
            EXACT_PENALIZED_NLL,
            EXP1_MAX_T_CRITICAL,
        ],
    }
)


FIXED_INPUTS.to_csv(
    OUTPUT_DIR
    / "experiment2_fixed_inputs.csv",
    index=False,
)


GRADIENT_VALIDATION.to_csv(
    OUTPUT_DIR
    / "experiment2_gradient_validation.csv",
    index=False,
)


ATTRIBUTION_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment2_descriptor_attribution.csv",
    index=False,
)


TAIL_ATTRIBUTION.to_csv(
    OUTPUT_DIR
    / "experiment2_tail_attribution.csv",
    index=False,
)


CURRENT_TAIL_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment2_tail_summary.csv",
    index=False,
)


SURVEILLANCE_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment2_surveillance_priorities.csv",
    index=False,
)


DESCRIPTOR_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment2_descriptor_summary.csv",
    index=False,
)


np.savetxt(
    OUTPUT_DIR
    / "experiment2_fixed_Sigma_eta.csv",
    SIGMA_ETA,
    delimiter=",",
)


np.savetxt(
    OUTPUT_DIR
    / "experiment2_fixed_Hessian_eta.csv",
    HESSIAN_ETA,
    delimiter=",",
)


np.savetxt(
    OUTPUT_DIR
    / "experiment2_fixed_Sigma_log_b_log_gamma.csv",
    SIGMA_BG,
    delimiter=",",
)


np.savetxt(
    OUTPUT_DIR
    / "experiment2_recursive_gradient_matrix.csv",
    RECURSIVE_GRADIENT,
    delimiter=",",
)


np.savetxt(
    OUTPUT_DIR
    / "experiment2_descriptor_covariance.csv",
    DESCRIPTOR_COVARIANCE,
    delimiter=",",
)


pd.DataFrame(
    {
        "new_infections":
            np.arange(
                len(
                    ADDITIONAL[
                        "pmf"
                    ]
                )
            ),

        "probability":
            ADDITIONAL[
                "pmf"
            ],

        "strict_tail_probability":
            ADDITIONAL[
                "strict_tail"
            ],
    }
).to_csv(
    OUTPUT_DIR
    / "experiment2_final_size_distribution.csv",
    index=False,
)


# ============================================================
# 18. FINAL DIAGNOSTICS
# ============================================================

TOTAL_RUNTIME = (
    time.perf_counter()
    - TOTAL_START
)


MAX_ABSOLUTE_GRADIENT_ERROR = float(
    np.max(
        GRADIENT_ERROR
    )
)


print()
print(
    "Final Experiment 2 diagnostics"
)

print(
    {
        "max_absolute_gradient_error":
            MAX_ABSOLUTE_GRADIENT_ERROR,

        "Sigma_BG_eigenvalues":
            BG_EIGENVALUES.tolist(),

        "dominant_parameter_uncertainty_direction":
            DOMINANT_DIRECTION.tolist(),

        "expected_final_size":
            FINAL_SIZE_MEAN,

        "expected_extinction_time":
            EXTINCTION_MEAN,

        "runtime_seconds":
            TOTAL_RUNTIME,
    }
)


print()
print(
    "Experiment 2 completed"
)

print(
    "Results saved in",
    OUTPUT_DIR,
)
```
