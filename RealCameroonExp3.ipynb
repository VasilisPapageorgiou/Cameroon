```python
# ============================================================
# EXPERIMENT 3
#
# LOCAL NEAR-BOUNDARY INTERVENTION SCREENING
#
# Purpose
#
# Compare
#
# 1. Plug-in dominance
# 2. Uncertainty-aware global one-sided max-t dominance
#
# The experiment is anchored at the fitted Cameroon model
# from Experiment 1.
#
# IMPORTANT
#
# Multiplicative interventions were replaced by smooth
# additive rate interventions.
#
# Under multiplicative interventions in the single-regime
# SIR model, final-size risk depends on b/gamma and policy
# ordering is essentially invariant to the common baseline
# parameter uncertainty.
#
# The additive intervention maps below create a genuine
# near-boundary decision problem while retaining the same
# fitted baseline model and final-size risk functional.
#
# No data loading
# No parameter fitting
# No Hessian estimation
#
# One Google Colab cell
# ============================================================


from __future__ import annotations

import time
from itertools import combinations
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from IPython.display import display


TOTAL_START = time.perf_counter()


# ============================================================
# 1. FIXED CAMEROON FIT FROM EXPERIMENT 1
# ============================================================

N = 175
S0 = 114
I0 = 1
R0_INITIAL = 60

assert N == S0 + I0 + R0_INITIAL


B_HAT = 0.7650254997116541
GAMMA_HAT = 0.23768509187018172
PHI_HAT = 2.4906108661339097


ETA_TRUE = np.log(
    [
        B_HAT,
        GAMMA_HAT,
    ]
)


SIGMA_ETA = np.array(
    [
        [
            0.01078625,
            0.01032878,
            -0.01802093,
        ],
        [
            0.01032878,
            0.01670211,
            -0.03299136,
        ],
        [
            -0.01802093,
            -0.03299136,
            0.55918411,
        ],
    ],
    dtype=float,
)


SIGMA_BG = SIGMA_ETA[
    np.ix_(
        [0, 1],
        [0, 1],
    )
]


# ============================================================
# 2. FINAL-SIZE THRESHOLDS
# ============================================================

THRESHOLD_FRACTIONS = np.array(
    [
        0.50,
        0.60,
        0.70,
        0.80,
        0.90,
    ],
    dtype=float,
)


THRESHOLDS = np.round(
    THRESHOLD_FRACTIONS
    * S0
).astype(int)


assert np.array_equal(
    THRESHOLDS,
    np.array(
        [
            57,
            68,
            80,
            91,
            103,
        ]
    ),
)


# Experiment 1 baseline risks used as a check

EXP1_TAIL_RISKS = np.array(
    [
        0.502185,
        0.492970,
        0.454210,
        0.318310,
        0.045407,
    ]
)


# ============================================================
# 3. MONTE CARLO SETTINGS
# ============================================================

ALPHA = 0.05

BASE_SEED = 20260807


# 500 is appropriate for manuscript results.

N_REPLICATIONS = 500


# Gaussian max-t draws within each replication.
#
# This is inexpensive because the dimension is only two.

N_MAXT_DRAWS = 999


# The experiment deliberately spans weak through strong
# information regimes.

INFORMATION_MULTIPLIERS = np.array(
    [
        0.25,
        0.50,
        1.00,
        2.00,
        4.00,
        8.00,
    ],
    dtype=float,
)


OUTPUT_DIR = Path(
    "/content/experiment3_near_boundary_screening"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ============================================================
# 4. LOCAL INTERVENTION DESIGN
#
# Transformation
#
# b_a     = b     - delta_b
# gamma_a = gamma + delta_gamma
#
# These are fixed absolute rate changes.
#
# They are calibrated once around the fitted Cameroon model.
# They are NOT changed after seeing the Monte Carlo results.
#
# Several policies are intentionally close in b_a/gamma_a.
# This creates the near-boundary comparison required to
# evaluate false elimination.
# ============================================================

POLICIES = {

    "No additional intervention": {
        "delta_b": 0.000,
        "delta_gamma": 0.000,
    },


    # Close candidate policies

    "Contact-focused response": {
        "delta_b": 0.100,
        "delta_gamma": 0.000,
    },

    "Recovery-focused response": {
        "delta_b": 0.000,
        "delta_gamma": 0.040,
    },

    "Balanced response A": {
        "delta_b": 0.050,
        "delta_gamma": 0.020,
    },

    "Balanced response B": {
        "delta_b": 0.070,
        "delta_gamma": 0.012,
    },


    # A more clearly dominated policy provides information
    # about screening power.

    "Weak response": {
        "delta_b": 0.030,
        "delta_gamma": 0.005,
    },
}


BASELINE_POLICY = (
    "No additional intervention"
)


CANDIDATE_POLICIES = [
    name
    for name in POLICIES
    if name != BASELINE_POLICY
]


# ============================================================
# 5. EXACT FINAL-SIZE RECURSION WITH SENSITIVITIES
#
# Local derivatives are with respect to
#
# log b_policy
# log gamma_policy
# ============================================================

def final_size_pmf_gradient(
    b: float,
    gamma: float,
):

    probability = np.zeros(
        I0 + 1,
        dtype=float,
    )

    probability[I0] = 1.0


    gradient = np.zeros(
        (
            2,
            I0 + 1,
        ),
        dtype=float,
    )


    pmf = np.zeros(
        S0 + 1,
        dtype=float,
    )


    pmf_gradient = np.zeros(
        (
            2,
            S0 + 1,
        ),
        dtype=float,
    )


    for susceptible in range(
        S0,
        -1,
        -1,
    ):

        maximum_infectious = (
            I0
            + S0
            - susceptible
        )


        required_length = (
            maximum_infectious
            + 1
        )


        if len(
            probability
        ) < required_length:

            padding = (
                required_length
                - len(
                    probability
                )
            )


            probability = np.pad(
                probability,
                (
                    0,
                    padding,
                ),
            )


            gradient = np.pad(
                gradient,
                (
                    (0, 0),
                    (0, padding),
                ),
            )


        if susceptible > 0:

            infection_component = (
                b
                * susceptible
                / N
            )


            denominator = (
                infection_component
                + gamma
            )


            infection_probability = (
                infection_component
                / denominator
            )


            recovery_probability = (
                gamma
                / denominator
            )


            derivative_infection = np.array(
                [
                    infection_probability
                    * recovery_probability,

                    -infection_probability
                    * recovery_probability,
                ],
                dtype=float,
            )


            derivative_recovery = (
                -derivative_infection
            )


            next_probability = np.zeros(
                maximum_infectious
                + 2,
                dtype=float,
            )


            next_gradient = np.zeros(
                (
                    2,
                    maximum_infectious
                    + 2,
                ),
                dtype=float,
            )


        else:

            infection_probability = 0.0
            recovery_probability = 1.0

            derivative_infection = np.zeros(
                2
            )

            derivative_recovery = np.zeros(
                2
            )

            next_probability = None
            next_gradient = None


        for infectious in range(
            maximum_infectious,
            0,
            -1,
        ):

            mass = probability[
                infectious
            ]


            mass_gradient = gradient[
                :,
                infectious,
            ].copy()


            if (
                abs(mass) < 1e-300
                and np.all(
                    np.abs(
                        mass_gradient
                    )
                    < 1e-300
                )
            ):

                continue


            # Recovery

            probability[
                infectious - 1
            ] += (
                mass
                * recovery_probability
            )


            gradient[
                :,
                infectious - 1,
            ] += (

                mass_gradient
                * recovery_probability

                + mass
                * derivative_recovery
            )


            # Infection

            if susceptible > 0:

                next_probability[
                    infectious + 1
                ] += (
                    mass
                    * infection_probability
                )


                next_gradient[
                    :,
                    infectious + 1,
                ] += (

                    mass_gradient
                    * infection_probability

                    + mass
                    * derivative_infection
                )


        infections_so_far = (
            S0
            - susceptible
        )


        pmf[
            infections_so_far
        ] = (
            probability[0]
        )


        pmf_gradient[
            :,
            infections_so_far,
        ] = gradient[
            :,
            0,
        ]


        if susceptible > 0:

            probability = (
                next_probability
            )

            gradient = (
                next_gradient
            )


    normalizer = float(
        pmf.sum()
    )


    normalizer_gradient = (
        pmf_gradient.sum(
            axis=1
        )
    )


    pmf_normalized = (
        pmf
        / normalizer
    )


    gradient_normalized = (

        pmf_gradient
        * normalizer

        - pmf[
            None,
            :
        ]
        * normalizer_gradient[
            :,
            None,
        ]

    ) / (
        normalizer**2
    )


    return (
        pmf_normalized,
        gradient_normalized,
    )


# ============================================================
# 6. STRICT TAIL RISK
# ============================================================

def strict_tail_and_gradient(
    pmf,
    pmf_gradient,
):

    nonstrict_tail = np.cumsum(
        pmf[::-1]
    )[::-1]


    nonstrict_gradient = np.cumsum(
        pmf_gradient[
            :,
            ::-1,
        ],
        axis=1,
    )[
        :,
        ::-1,
    ]


    strict_tail = (
        nonstrict_tail
        - pmf
    )


    strict_gradient = (
        nonstrict_gradient
        - pmf_gradient
    )


    return (
        strict_tail,
        strict_gradient,
    )


# ============================================================
# 7. POLICY TRANSFORMATION AND CHAIN RULE
# ============================================================

def evaluate_policy(
    eta_bg,
    policy_name,
):

    baseline_b = float(
        np.exp(
            eta_bg[0]
        )
    )


    baseline_gamma = float(
        np.exp(
            eta_bg[1]
        )
    )


    specification = (
        POLICIES[
            policy_name
        ]
    )


    delta_b = specification[
        "delta_b"
    ]

    delta_gamma = specification[
        "delta_gamma"
    ]


    policy_b = (
        baseline_b
        - delta_b
    )


    policy_gamma = (
        baseline_gamma
        + delta_gamma
    )


    if policy_b <= 0.0:

        raise RuntimeError(
            "An intervention produced a nonpositive "
            "transmission rate."
        )


    (
        pmf,
        pmf_gradient_local,
    ) = final_size_pmf_gradient(
        b=policy_b,
        gamma=policy_gamma,
    )


    (
        tail,
        tail_gradient_local,
    ) = strict_tail_and_gradient(
        pmf,
        pmf_gradient_local,
    )


    risk_vector = (
        tail[
            THRESHOLDS
        ]
    )


    local_gradient = (
        tail_gradient_local[
            :,
            THRESHOLDS,
        ].T
    )


    # --------------------------------------------------------
    # Chain rule
    #
    # policy_b = b - delta_b
    #
    # d log(policy_b) / d log(b)
    #
    # = b / policy_b
    #
    # policy_gamma = gamma + delta_gamma
    #
    # d log(policy_gamma) / d log(gamma)
    #
    # = gamma / policy_gamma
    # --------------------------------------------------------

    chain_scale = np.array(
        [
            baseline_b
            / policy_b,

            baseline_gamma
            / policy_gamma,
        ],
        dtype=float,
    )


    risk_gradient = (
        local_gradient
        * chain_scale[
            None,
            :
        ]
    )


    return {

        "policy":
            policy_name,

        "b":
            policy_b,

        "gamma":
            policy_gamma,

        "ratio":
            policy_b
            / policy_gamma,

        "risk_vector":
            risk_vector,

        "risk_gradient":
            risk_gradient,
    }


def evaluate_all_policies(
    eta_bg,
):

    return {

        policy_name:
            evaluate_policy(
                eta_bg,
                policy_name,
            )

        for policy_name
        in POLICIES
    }


# ============================================================
# 8. ORACLE POLICY CURVES
# ============================================================

TRUE_POLICY_RESULTS = (
    evaluate_all_policies(
        ETA_TRUE
    )
)


# Baseline check against Experiment 1

baseline_difference = np.max(
    np.abs(
        TRUE_POLICY_RESULTS[
            BASELINE_POLICY
        ][
            "risk_vector"
        ]
        - EXP1_TAIL_RISKS
    )
)


print()
print(
    "Maximum baseline difference from Experiment 1"
)

print(
    baseline_difference
)


if baseline_difference > 5e-4:

    raise RuntimeError(
        "The final-size recursion does not reproduce "
        "Experiment 1."
    )


# ============================================================
# 9. POLICY DESIGN TABLE
# ============================================================

DESIGN_ROWS = []


for policy_name, result in (
    TRUE_POLICY_RESULTS.items()
):

    row = {

        "policy":
            policy_name,

        "b":
            result[
                "b"
            ],

        "gamma":
            result[
                "gamma"
            ],

        "b_over_gamma":
            result[
                "ratio"
            ],
    }


    for index, threshold in enumerate(
        THRESHOLDS
    ):

        row[
            f"risk_{threshold}"
        ] = (
            result[
                "risk_vector"
            ][
                index
            ]
        )


    DESIGN_ROWS.append(
        row
    )


POLICY_DESIGN = pd.DataFrame(
    DESIGN_ROWS
)


print()
print(
    "Local intervention design"
)

display(
    POLICY_DESIGN.round(
        5
    )
)


# ============================================================
# 10. PLUG-IN DOMINANCE RULE
# ============================================================

def dominates_vector(
    first,
    second,
    tolerance=1e-10,
):

    return bool(

        np.all(
            first
            <= second
            + tolerance
        )

        and np.any(
            first
            < second
            - tolerance
        )
    )


def plugin_gate(
    policy_names,
    policy_results,
):

    eliminated = set()

    rows = []


    for first in policy_names:

        for second in policy_names:

            if first == second:

                continue


            contrast = (

                policy_results[
                    first
                ][
                    "risk_vector"
                ]

                - policy_results[
                    second
                ][
                    "risk_vector"
                ]
            )


            dominance = (
                dominates_vector(

                    policy_results[
                        first
                    ][
                        "risk_vector"
                    ],

                    policy_results[
                        second
                    ][
                        "risk_vector"
                    ],
                )
            )


            if dominance:

                eliminated.add(
                    second
                )


            rows.append(
                {
                    "potential_dominator":
                        first,

                    "potentially_dominated":
                        second,

                    "maximum_contrast":
                        float(
                            np.max(
                                contrast
                            )
                        ),

                    "plugin_dominance":
                        dominance,
                }
            )


    retained = (
        set(
            policy_names
        )
        - eliminated
    )


    return (
        retained,
        pd.DataFrame(
            rows
        ),
    )


# ============================================================
# 11. GLOBAL ONE-SIDED MAX-T ROBUST GATE
# ============================================================

def robust_gate(
    policy_names,
    policy_results,
    covariance,
    rng,
    n_draws,
):

    parameter_draws = (
        rng.multivariate_normal(
            mean=np.zeros(
                2
            ),
            cov=covariance,
            size=n_draws,
        )
    )


    pair_information = []

    standardized_blocks = []


    for first in policy_names:

        for second in policy_names:

            if first == second:

                continue


            contrast = (

                policy_results[
                    first
                ][
                    "risk_vector"
                ]

                - policy_results[
                    second
                ][
                    "risk_vector"
                ]
            )


            contrast_gradient = (

                policy_results[
                    first
                ][
                    "risk_gradient"
                ]

                - policy_results[
                    second
                ][
                    "risk_gradient"
                ]
            )


            variance = np.maximum(
                np.einsum(
                    "ki,ij,kj->k",
                    contrast_gradient,
                    covariance,
                    contrast_gradient,
                ),
                0.0,
            )


            standard_error = np.sqrt(
                variance
            )


            active = (
                standard_error
                > 1e-12
            )


            perturbation = (

                parameter_draws

                @ contrast_gradient.T
            )


            if np.any(
                active
            ):

                standardized_blocks.append(

                    perturbation[
                        :,
                        active,
                    ]

                    / standard_error[
                        active
                    ]
                )


            pair_information.append(
                {
                    "first":
                        first,

                    "second":
                        second,

                    "contrast":
                        contrast,

                    "standard_error":
                        standard_error,
                }
            )


    if standardized_blocks:

        standardized_all = np.concatenate(
            standardized_blocks,
            axis=1,
        )


        max_statistic = np.max(
            standardized_all,
            axis=1,
        )


        critical_value = float(
            np.quantile(
                max_statistic,
                1.0 - ALPHA,
            )
        )


    else:

        critical_value = 0.0


    eliminated = set()

    rows = []


    for information in (
        pair_information
    ):

        contrast = information[
            "contrast"
        ]


        standard_error = information[
            "standard_error"
        ]


        upper_band = (

            contrast

            + critical_value
            * standard_error
        )


        dominance = bool(

            np.all(
                upper_band
                <= 1e-10
            )

            and np.any(
                upper_band
                < -1e-10
            )
        )


        if dominance:

            eliminated.add(
                information[
                    "second"
                ]
            )


        rows.append(
            {
                "potential_dominator":
                    information[
                        "first"
                    ],

                "potentially_dominated":
                    information[
                        "second"
                    ],

                "maximum_contrast":
                    float(
                        np.max(
                            contrast
                        )
                    ),

                "maximum_upper_band":
                    float(
                        np.max(
                            upper_band
                        )
                    ),

                "robust_dominance":
                    dominance,
            }
        )


    retained = (
        set(
            policy_names
        )
        - eliminated
    )


    return (
        retained,
        pd.DataFrame(
            rows
        ),
        critical_value,
    )


# ============================================================
# 12. AMBIGUITY
# ============================================================

def ambiguity_rate(
    policy_names,
    pairwise_table,
    dominance_column,
):

    values = []


    for first, second in combinations(
        policy_names,
        2,
    ):

        first_dominates = bool(

            pairwise_table.loc[
                (
                    pairwise_table[
                        "potential_dominator"
                    ] == first
                )
                & (
                    pairwise_table[
                        "potentially_dominated"
                    ] == second
                ),
                dominance_column,
            ].iloc[0]
        )


        second_dominates = bool(

            pairwise_table.loc[
                (
                    pairwise_table[
                        "potential_dominator"
                    ] == second
                )
                & (
                    pairwise_table[
                        "potentially_dominated"
                    ] == first
                ),
                dominance_column,
            ].iloc[0]
        )


        values.append(

            not (
                first_dominates
                or second_dominates
            )
        )


    return float(
        np.mean(
            values
        )
    )


# ============================================================
# 13. ORACLE DOMINANCE STRUCTURE
# ============================================================

(
    ORACLE_RETAINED,
    ORACLE_PAIRWISE,
) = plugin_gate(
    CANDIDATE_POLICIES,
    TRUE_POLICY_RESULTS,
)


ORACLE_DOMINATED = (

    set(
        CANDIDATE_POLICIES
    )

    - ORACLE_RETAINED
)


print()
print(
    "Oracle retained intervention set"
)

print(
    sorted(
        ORACLE_RETAINED
    )
)


print()
print(
    "Oracle dominated interventions"
)

print(
    sorted(
        ORACLE_DOMINATED
    )
)


# ============================================================
# 14. FITTED CAMEROON GATE
# ============================================================

APPLICATION_RNG = np.random.default_rng(
    BASE_SEED
)


(
    APPLICATION_PLUGIN_RETAINED,
    APPLICATION_PLUGIN_TABLE,
) = plugin_gate(
    CANDIDATE_POLICIES,
    TRUE_POLICY_RESULTS,
)


(
    APPLICATION_ROBUST_RETAINED,
    APPLICATION_ROBUST_TABLE,
    APPLICATION_CRITICAL_VALUE,
) = robust_gate(
    CANDIDATE_POLICIES,
    TRUE_POLICY_RESULTS,
    SIGMA_BG,
    APPLICATION_RNG,
    10000,
)


print()
print(
    "Plug-in retained set at fitted Cameroon parameters"
)

print(
    sorted(
        APPLICATION_PLUGIN_RETAINED
    )
)


print()
print(
    "Robust retained set at fitted Cameroon parameters"
)

print(
    sorted(
        APPLICATION_ROBUST_RETAINED
    )
)


print()
print(
    "Global one-sided max-t critical value"
)

print(
    APPLICATION_CRITICAL_VALUE
)


# ============================================================
# 15. REPEATED-SAMPLING STRESS TEST
#
# eta_hat is generated around the fitted Cameroon parameter.
#
# The intervention definitions remain fixed.
#
# The only quantity changed across information regimes is
# estimator covariance.
# ============================================================

SIMULATION_START = time.perf_counter()


SIMULATION_RNG = np.random.default_rng(
    BASE_SEED
    + 10000
)


SIMULATION_ROWS = []


for information_multiplier in (
    INFORMATION_MULTIPLIERS
):

    covariance = (

        SIGMA_BG

        / information_multiplier
    )


    print()
    print(
        "Relative information",
        information_multiplier,
    )


    for replication in range(
        N_REPLICATIONS
    ):

        eta_estimate = (
            SIMULATION_RNG.multivariate_normal(
                mean=
                    ETA_TRUE,
                cov=
                    covariance,
            )
        )


        estimated_results = (
            evaluate_all_policies(
                eta_estimate
            )
        )


        # ----------------------------------------------------
        # Plug-in
        # ----------------------------------------------------

        (
            plugin_retained,
            plugin_table,
        ) = plugin_gate(
            CANDIDATE_POLICIES,
            estimated_results,
        )


        plugin_eliminated = (

            set(
                CANDIDATE_POLICIES
            )

            - plugin_retained
        )


        plugin_ambiguity = (
            ambiguity_rate(
                CANDIDATE_POLICIES,
                plugin_table,
                "plugin_dominance",
            )
        )


        # ----------------------------------------------------
        # Robust
        # ----------------------------------------------------

        (
            robust_retained,
            robust_table,
            robust_critical_value,
        ) = robust_gate(
            CANDIDATE_POLICIES,
            estimated_results,
            covariance,
            SIMULATION_RNG,
            N_MAXT_DRAWS,
        )


        robust_eliminated = (

            set(
                CANDIDATE_POLICIES
            )

            - robust_retained
        )


        robust_ambiguity = (
            ambiguity_rate(
                CANDIDATE_POLICIES,
                robust_table,
                "robust_dominance",
            )
        )


        # ----------------------------------------------------
        # Metrics
        # ----------------------------------------------------

        for (
            method,
            retained,
            eliminated,
            ambiguity,
        ) in [

            (
                "Plug-in gate",
                plugin_retained,
                plugin_eliminated,
                plugin_ambiguity,
            ),

            (
                "Robust simultaneous gate",
                robust_retained,
                robust_eliminated,
                robust_ambiguity,
            ),
        ]:


            false_elimination = (

                len(
                    eliminated
                    & ORACLE_RETAINED
                )

                > 0
            )


            screening_power = (

                len(
                    eliminated
                    & ORACLE_DOMINATED
                )

                / len(
                    ORACLE_DOMINATED
                )
            )


            exact_oracle_set = (

                retained
                == ORACLE_RETAINED
            )


            SIMULATION_ROWS.append(
                {
                    "information_multiplier":
                        information_multiplier,

                    "replication":
                        replication + 1,

                    "method":
                        method,

                    "false_elimination":
                        float(
                            false_elimination
                        ),

                    "screening_power":
                        screening_power,

                    "retained_set_size":
                        len(
                            retained
                        ),

                    "oracle_set_recovery":
                        float(
                            exact_oracle_set
                        ),

                    "ambiguity_rate":
                        ambiguity,

                    "critical_value":
                        (
                            robust_critical_value
                            if method
                            == "Robust simultaneous gate"
                            else np.nan
                        ),
                }
            )


        if (
            replication + 1
        ) % 100 == 0:

            print(
                "Replication",
                replication + 1,
                "of",
                N_REPLICATIONS,
            )


SIMULATION_RESULTS = pd.DataFrame(
    SIMULATION_ROWS
)


SIMULATION_RUNTIME = (
    time.perf_counter()
    - SIMULATION_START
)


# ============================================================
# 16. MONTE CARLO SUMMARIES
# ============================================================

def summarize_metric(
    values,
):

    values = np.asarray(
        values,
        dtype=float,
    )


    values = values[
        np.isfinite(
            values
        )
    ]


    mean = float(
        np.mean(
            values
        )
    )


    mcse = float(

        np.std(
            values,
            ddof=1,
        )

        / np.sqrt(
            len(
                values
            )
        )
    )


    return (
        mean,
        mcse,
    )


SUMMARY_ROWS = []


for (
    information_multiplier,
    method
), group in (
    SIMULATION_RESULTS.groupby(
        [
            "information_multiplier",
            "method",
        ]
    )
):


    false_mean, false_mcse = (
        summarize_metric(
            group[
                "false_elimination"
            ]
        )
    )


    power_mean, power_mcse = (
        summarize_metric(
            group[
                "screening_power"
            ]
        )
    )


    retained_mean, retained_mcse = (
        summarize_metric(
            group[
                "retained_set_size"
            ]
        )
    )


    oracle_mean, oracle_mcse = (
        summarize_metric(
            group[
                "oracle_set_recovery"
            ]
        )
    )


    ambiguity_mean, ambiguity_mcse = (
        summarize_metric(
            group[
                "ambiguity_rate"
            ]
        )
    )


    SUMMARY_ROWS.append(
        {
            "information_multiplier":
                information_multiplier,

            "method":
                method,

            "false_elimination_probability":
                false_mean,

            "false_elimination_mcse":
                false_mcse,

            "screening_power":
                power_mean,

            "screening_power_mcse":
                power_mcse,

            "mean_retained_set_size":
                retained_mean,

            "retained_set_size_mcse":
                retained_mcse,

            "oracle_set_recovery":
                oracle_mean,

            "oracle_set_recovery_mcse":
                oracle_mcse,

            "ambiguity_rate":
                ambiguity_mean,

            "ambiguity_rate_mcse":
                ambiguity_mcse,
        }
    )


SCREENING_SUMMARY = pd.DataFrame(
    SUMMARY_ROWS
)


SCREENING_SUMMARY = (
    SCREENING_SUMMARY
    .sort_values(
        [
            "information_multiplier",
            "method",
        ]
    )
    .reset_index(
        drop=True
    )
)


print()
print(
    "Repeated-sampling screening results"
)

display(
    SCREENING_SUMMARY.round(
        4
    )
)


# ============================================================
# 17. DIRECT METHOD COMPARISON
# ============================================================

plugin_summary = (
    SCREENING_SUMMARY.loc[
        SCREENING_SUMMARY[
            "method"
        ]
        == "Plug-in gate"
    ]
    .set_index(
        "information_multiplier"
    )
)


robust_summary = (
    SCREENING_SUMMARY.loc[
        SCREENING_SUMMARY[
            "method"
        ]
        == "Robust simultaneous gate"
    ]
    .set_index(
        "information_multiplier"
    )
)


COMPARISON_SUMMARY = pd.DataFrame(
    {
        "information_multiplier":
            INFORMATION_MULTIPLIERS,

        "plugin_false_elimination":
            plugin_summary.loc[
                INFORMATION_MULTIPLIERS,
                "false_elimination_probability",
            ].to_numpy(),

        "robust_false_elimination":
            robust_summary.loc[
                INFORMATION_MULTIPLIERS,
                "false_elimination_probability",
            ].to_numpy(),

        "false_elimination_reduction_percentage_points":
            100.0
            * (
                plugin_summary.loc[
                    INFORMATION_MULTIPLIERS,
                    "false_elimination_probability",
                ].to_numpy()

                - robust_summary.loc[
                    INFORMATION_MULTIPLIERS,
                    "false_elimination_probability",
                ].to_numpy()
            ),

        "plugin_screening_power":
            plugin_summary.loc[
                INFORMATION_MULTIPLIERS,
                "screening_power",
            ].to_numpy(),

        "robust_screening_power":
            robust_summary.loc[
                INFORMATION_MULTIPLIERS,
                "screening_power",
            ].to_numpy(),

        "plugin_oracle_recovery":
            plugin_summary.loc[
                INFORMATION_MULTIPLIERS,
                "oracle_set_recovery",
            ].to_numpy(),

        "robust_oracle_recovery":
            robust_summary.loc[
                INFORMATION_MULTIPLIERS,
                "oracle_set_recovery",
            ].to_numpy(),
    }
)


print()
print(
    "Direct plug-in versus robust comparison"
)

display(
    COMPARISON_SUMMARY.round(
        4
    )
)


# ============================================================
# 18. FIGURE 1
#
# Local risk curves and false elimination
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        13,
        5,
    ),
)


# ------------------------------------------------------------
# Local intervention curves
# ------------------------------------------------------------

for policy_name in (
    POLICIES
):

    result = (
        TRUE_POLICY_RESULTS[
            policy_name
        ]
    )


    linewidth = (
        2.8
        if policy_name
        == BASELINE_POLICY
        else 1.8
    )


    axes[0].plot(
        THRESHOLDS,
        result[
            "risk_vector"
        ],
        marker="o",
        linewidth=
            linewidth,
        label=
            policy_name,
    )


axes[0].set_xlabel(
    "Final-size threshold"
)

axes[0].set_ylabel(
    r"$\Pr(N^I>c)$"
)

axes[0].set_ylim(
    -0.01,
    0.60,
)

axes[0].grid(
    alpha=0.20
)

axes[0].legend(
    frameon=False,
    fontsize=7,
)


# ------------------------------------------------------------
# False elimination
# ------------------------------------------------------------

for method, group in (
    SCREENING_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    axes[1].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "false_elimination_probability"
        ],
        yerr=
            1.96
            * group[
                "false_elimination_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=4,
        label=
            method,
    )


axes[1].axhline(
    ALPHA,
    color="black",
    linestyle="--",
    linewidth=1.2,
    label="0.05 reference",
)


axes[1].set_xlabel(
    "Relative information"
)

axes[1].set_ylabel(
    "False-elimination probability"
)

axes[1].set_ylim(
    -0.01,
    0.65,
)

axes[1].grid(
    alpha=0.20
)

axes[1].legend(
    frameon=False
)


fig.tight_layout()


fig.savefig(
    OUTPUT_DIR
    / "experiment3_false_elimination.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 19. FIGURE 2
#
# Protection versus screening power
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(
        15,
        4.6,
    ),
)


for method, group in (
    SCREENING_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    axes[0].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "screening_power"
        ],
        yerr=
            1.96
            * group[
                "screening_power_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method,
    )


    axes[1].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "mean_retained_set_size"
        ],
        yerr=
            1.96
            * group[
                "retained_set_size_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method,
    )


    axes[2].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "oracle_set_recovery"
        ],
        yerr=
            1.96
            * group[
                "oracle_set_recovery_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method,
    )


axes[0].set_ylabel(
    "Screening power"
)

axes[1].set_ylabel(
    "Mean retained-set size"
)

axes[2].set_ylabel(
    "Oracle-set recovery probability"
)


for axis in axes:

    axis.set_xlabel(
        "Relative information"
    )

    axis.grid(
        alpha=0.20
    )

    axis.legend(
        frameon=False,
        fontsize=8,
    )


axes[0].set_ylim(
    -0.01,
    1.01,
)

axes[2].set_ylim(
    -0.01,
    1.01,
)


fig.tight_layout()


fig.savefig(
    OUTPUT_DIR
    / "experiment3_power_and_stability.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 20. SAVE RESULTS
# ============================================================

POLICY_DESIGN.to_csv(
    OUTPUT_DIR
    / "experiment3_policy_design.csv",
    index=False,
)


ORACLE_PAIRWISE.to_csv(
    OUTPUT_DIR
    / "experiment3_oracle_pairwise.csv",
    index=False,
)


APPLICATION_PLUGIN_TABLE.to_csv(
    OUTPUT_DIR
    / "experiment3_application_plugin_gate.csv",
    index=False,
)


APPLICATION_ROBUST_TABLE.to_csv(
    OUTPUT_DIR
    / "experiment3_application_robust_gate.csv",
    index=False,
)


SIMULATION_RESULTS.to_csv(
    OUTPUT_DIR
    / "experiment3_replication_results.csv",
    index=False,
)


SCREENING_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_screening_summary.csv",
    index=False,
)


COMPARISON_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_method_comparison.csv",
    index=False,
)


np.savetxt(
    OUTPUT_DIR
    / "experiment3_Sigma_log_b_log_gamma.csv",
    SIGMA_BG,
    delimiter=",",
)


# ============================================================
# 21. FINAL DIAGNOSTICS
# ============================================================

TOTAL_RUNTIME = (
    time.perf_counter()
    - TOTAL_START
)


print()
print(
    "Experiment 3 completed"
)

print(
    {
        "oracle_retained":
            sorted(
                ORACLE_RETAINED
            ),

        "number_oracle_dominated":
            len(
                ORACLE_DOMINATED
            ),

        "application_plugin_retained":
            sorted(
                APPLICATION_PLUGIN_RETAINED
            ),

        "application_robust_retained":
            sorted(
                APPLICATION_ROBUST_RETAINED
            ),

        "runtime_seconds":
            TOTAL_RUNTIME,
    }
)


print()
print(
    "Results saved in",
    OUTPUT_DIR,
)
```

```python
# ============================================================
# EXPERIMENT 3B
#
# STRATIFIED INTERVENTION SCREENING
#
# Run AFTER the current Experiment 3 cell.
#
# Purpose
#
# The experiment now contains two prespecified classes
#
# 1. Near-boundary interventions
#    These test protection against false elimination.
#
# 2. Separated dominated interventions
#    These test whether the robust rule retains useful
#    screening power when dominance is sufficiently clear.
#
# The fitted Cameroon parameters, covariance matrix,
# thresholds, CTMC final-size recursion, and screening
# methodology are NOT changed.
# ============================================================


from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from IPython.display import display


# ============================================================
# 1. CHECK THAT EXPERIMENT 3 HAS ALREADY BEEN RUN
# ============================================================

_REQUIRED_OBJECTS = [
    "ETA_TRUE",
    "SIGMA_BG",
    "THRESHOLDS",
    "evaluate_all_policies",
    "plugin_gate",
    "robust_gate",
    "ALPHA",
    "BASE_SEED",
]

for object_name in _REQUIRED_OBJECTS:

    if object_name not in globals():

        raise RuntimeError(
            "Run the current Experiment 3 cell first. "
            f"Missing object {object_name}."
        )


# ============================================================
# 2. REVISED PRESPECIFIED INTERVENTION DESIGN
#
# Baseline and near-boundary interventions are retained.
#
# Three additional separated interventions are included.
#
# They are deliberately weaker than the near-boundary
# candidates so that the experiment contains both
#
# difficult comparisons
#
# and
#
# clearly separated comparisons.
#
# This is a design feature of the simulation and must be
# reported as such.
# ============================================================

POLICIES = {

    "No additional intervention": {

        "delta_b":
            0.000,

        "delta_gamma":
            0.000,

        "decision_class":
            "Reference",
    },


    # --------------------------------------------------------
    # Near-boundary candidates
    # --------------------------------------------------------

    "Contact-focused response": {

        "delta_b":
            0.100,

        "delta_gamma":
            0.000,

        "decision_class":
            "Near-boundary",
    },


    "Recovery-focused response": {

        "delta_b":
            0.000,

        "delta_gamma":
            0.040,

        "decision_class":
            "Near-boundary",
    },


    "Balanced response A": {

        "delta_b":
            0.050,

        "delta_gamma":
            0.020,

        "decision_class":
            "Near-boundary",
    },


    "Balanced response B": {

        "delta_b":
            0.070,

        "delta_gamma":
            0.012,

        "decision_class":
            "Near-boundary",
    },


    # --------------------------------------------------------
    # Separated dominated candidates
    #
    # These remain beneficial relative to no intervention
    # but are appreciably weaker than the best near-boundary
    # alternatives.
    # --------------------------------------------------------

    "Weak balanced response": {

        "delta_b":
            0.045,

        "delta_gamma":
            0.005,

        "decision_class":
            "Separated",
    },


    "Very weak response": {

        "delta_b":
            0.025,

        "delta_gamma":
            0.004,

        "decision_class":
            "Separated",
    },


    "Minimal response": {

        "delta_b":
            0.010,

        "delta_gamma":
            0.002,

        "decision_class":
            "Separated",
    },
}


BASELINE_POLICY = (
    "No additional intervention"
)


CANDIDATE_POLICIES = [

    policy_name

    for policy_name in POLICIES

    if policy_name
    != BASELINE_POLICY
]


NEAR_BOUNDARY_POLICIES = {

    policy_name

    for policy_name, specification
    in POLICIES.items()

    if specification[
        "decision_class"
    ] == "Near-boundary"
}


SEPARATED_POLICIES = {

    policy_name

    for policy_name, specification
    in POLICIES.items()

    if specification[
        "decision_class"
    ] == "Separated"
}


# ============================================================
# 3. RECOMPUTE POLICY CURVES AT THE FIXED CAMEROON FIT
# ============================================================

TRUE_POLICY_RESULTS = (
    evaluate_all_policies(
        ETA_TRUE
    )
)


# ============================================================
# 4. POLICY DESIGN TABLE
# ============================================================

POLICY_DESIGN_ROWS = []


for policy_name, result in (
    TRUE_POLICY_RESULTS.items()
):

    specification = (
        POLICIES[
            policy_name
        ]
    )


    row = {

        "policy":
            policy_name,

        "decision_class":
            specification[
                "decision_class"
            ],

        "delta_b":
            specification[
                "delta_b"
            ],

        "delta_gamma":
            specification[
                "delta_gamma"
            ],

        "resulting_b":
            result[
                "b"
            ],

        "resulting_gamma":
            result[
                "gamma"
            ],

        "b_over_gamma":
            result[
                "ratio"
            ],
    }


    for threshold_index, threshold in enumerate(
        THRESHOLDS
    ):

        row[
            f"risk_{threshold}"
        ] = (
            result[
                "risk_vector"
            ][
                threshold_index
            ]
        )


    POLICY_DESIGN_ROWS.append(
        row
    )


STRATIFIED_POLICY_DESIGN = pd.DataFrame(
    POLICY_DESIGN_ROWS
)


print()
print(
    "Revised policy design"
)

display(
    STRATIFIED_POLICY_DESIGN.round(
        5
    )
)


# ============================================================
# 5. ORACLE DOMINANCE STRUCTURE
# ============================================================

(
    ORACLE_RETAINED,
    ORACLE_PAIRWISE,
) = plugin_gate(
    CANDIDATE_POLICIES,
    TRUE_POLICY_RESULTS,
)


ORACLE_DOMINATED = (

    set(
        CANDIDATE_POLICIES
    )

    - ORACLE_RETAINED
)


ORACLE_DOMINATED_NEAR = (

    ORACLE_DOMINATED

    & NEAR_BOUNDARY_POLICIES
)


ORACLE_DOMINATED_SEPARATED = (

    ORACLE_DOMINATED

    & SEPARATED_POLICIES
)


print()
print(
    "Oracle retained set"
)

print(
    sorted(
        ORACLE_RETAINED
    )
)


print()
print(
    "Oracle dominated near-boundary policies"
)

print(
    sorted(
        ORACLE_DOMINATED_NEAR
    )
)


print()
print(
    "Oracle dominated separated policies"
)

print(
    sorted(
        ORACLE_DOMINATED_SEPARATED
    )
)


if len(
    ORACLE_DOMINATED_NEAR
) == 0:

    raise RuntimeError(
        "No oracle-dominated near-boundary "
        "interventions were generated."
    )


if len(
    ORACLE_DOMINATED_SEPARATED
) == 0:

    raise RuntimeError(
        "No oracle-dominated separated "
        "interventions were generated."
    )


# ============================================================
# 6. MONTE CARLO SETTINGS
#
# 500 is enough for development.
#
# For the final paper use
#
# N_REPLICATIONS_STRATIFIED = 1000
# ============================================================

N_REPLICATIONS_STRATIFIED = 500

N_MAXT_DRAWS_STRATIFIED = 999


INFORMATION_MULTIPLIERS_STRATIFIED = np.array(
    [
        0.25,
        0.50,
        1.00,
        2.00,
        4.00,
        8.00,
    ],
    dtype=float,
)


STRATIFIED_OUTPUT_DIR = Path(
    "/content/experiment3_stratified_screening"
)


STRATIFIED_OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ============================================================
# 7. MONTE CARLO EXPERIMENT
# ============================================================

STRATIFIED_RNG = np.random.default_rng(
    BASE_SEED
    + 30000
)


STRATIFIED_ROWS = []


for information_multiplier in (
    INFORMATION_MULTIPLIERS_STRATIFIED
):

    covariance = (

        SIGMA_BG

        / information_multiplier
    )


    print()
    print(
        "Relative information",
        information_multiplier
    )


    for replication in range(
        N_REPLICATIONS_STRATIFIED
    ):

        # ----------------------------------------------------
        # Working estimator distribution around the fixed
        # Cameroon parameter vector
        # ----------------------------------------------------

        eta_estimate = (
            STRATIFIED_RNG.multivariate_normal(
                mean=
                    ETA_TRUE,
                cov=
                    covariance,
            )
        )


        estimated_results = (
            evaluate_all_policies(
                eta_estimate
            )
        )


        # ----------------------------------------------------
        # Plug-in gate
        # ----------------------------------------------------

        (
            plugin_retained,
            plugin_table,
        ) = plugin_gate(
            CANDIDATE_POLICIES,
            estimated_results,
        )


        plugin_eliminated = (

            set(
                CANDIDATE_POLICIES
            )

            - plugin_retained
        )


        # ----------------------------------------------------
        # Robust simultaneous gate
        # ----------------------------------------------------

        (
            robust_retained,
            robust_table,
            robust_critical_value,
        ) = robust_gate(
            CANDIDATE_POLICIES,
            estimated_results,
            covariance,
            STRATIFIED_RNG,
            N_MAXT_DRAWS_STRATIFIED,
        )


        robust_eliminated = (

            set(
                CANDIDATE_POLICIES
            )

            - robust_retained
        )


        # ----------------------------------------------------
        # Method-specific metrics
        # ----------------------------------------------------

        for (
            method_name,
            retained_set,
            eliminated_set,
        ) in [

            (
                "Plug-in gate",
                plugin_retained,
                plugin_eliminated,
            ),

            (
                "Robust simultaneous gate",
                robust_retained,
                robust_eliminated,
            ),
        ]:


            # ------------------------------------------------
            # Familywise false elimination
            #
            # At least one truly retained intervention
            # has been incorrectly eliminated.
            # ------------------------------------------------

            false_elimination = float(

                len(
                    eliminated_set
                    & ORACLE_RETAINED
                )

                > 0
            )


            # ------------------------------------------------
            # Overall screening power
            # ------------------------------------------------

            overall_power = (

                len(
                    eliminated_set
                    & ORACLE_DOMINATED
                )

                / len(
                    ORACLE_DOMINATED
                )
            )


            # ------------------------------------------------
            # Near-boundary screening power
            # ------------------------------------------------

            near_boundary_power = (

                len(
                    eliminated_set
                    & ORACLE_DOMINATED_NEAR
                )

                / len(
                    ORACLE_DOMINATED_NEAR
                )
            )


            # ------------------------------------------------
            # Separated-alternative screening power
            # ------------------------------------------------

            separated_power = (

                len(
                    eliminated_set
                    & ORACLE_DOMINATED_SEPARATED
                )

                / len(
                    ORACLE_DOMINATED_SEPARATED
                )
            )


            # ------------------------------------------------
            # Exact oracle-set recovery
            # ------------------------------------------------

            oracle_recovery = float(

                retained_set
                == ORACLE_RETAINED
            )


            STRATIFIED_ROWS.append(
                {

                    "information_multiplier":
                        information_multiplier,

                    "replication":
                        replication + 1,

                    "method":
                        method_name,

                    "false_elimination":
                        false_elimination,

                    "overall_screening_power":
                        overall_power,

                    "near_boundary_power":
                        near_boundary_power,

                    "separated_power":
                        separated_power,

                    "retained_set_size":
                        len(
                            retained_set
                        ),

                    "oracle_set_recovery":
                        oracle_recovery,

                    "critical_value":
                        (
                            robust_critical_value

                            if method_name
                            == "Robust simultaneous gate"

                            else np.nan
                        ),
                }
            )


        if (
            replication + 1
        ) % 100 == 0:

            print(
                "Replication",
                replication + 1,
                "of",
                N_REPLICATIONS_STRATIFIED,
            )


STRATIFIED_RESULTS = pd.DataFrame(
    STRATIFIED_ROWS
)


# ============================================================
# 8. MONTE CARLO SUMMARY
# ============================================================

def summarize_mean_mcse(
    series,
):

    values = np.asarray(
        series,
        dtype=float,
    )


    values = values[
        np.isfinite(
            values
        )
    ]


    mean_value = float(
        np.mean(
            values
        )
    )


    mcse_value = float(

        np.std(
            values,
            ddof=1,
        )

        / np.sqrt(
            len(
                values
            )
        )
    )


    return (
        mean_value,
        mcse_value,
    )


SUMMARY_ROWS = []


for (
    information_multiplier,
    method_name
), group in (
    STRATIFIED_RESULTS.groupby(
        [
            "information_multiplier",
            "method",
        ]
    )
):

    row = {

        "information_multiplier":
            information_multiplier,

        "method":
            method_name,
    }


    for metric in [

        "false_elimination",

        "overall_screening_power",

        "near_boundary_power",

        "separated_power",

        "retained_set_size",

        "oracle_set_recovery",
    ]:

        mean_value, mcse_value = (
            summarize_mean_mcse(
                group[
                    metric
                ]
            )
        )


        row[
            metric
        ] = (
            mean_value
        )


        row[
            metric
            + "_mcse"
        ] = (
            mcse_value
        )


    SUMMARY_ROWS.append(
        row
    )


STRATIFIED_SUMMARY = pd.DataFrame(
    SUMMARY_ROWS
)


STRATIFIED_SUMMARY = (
    STRATIFIED_SUMMARY
    .sort_values(
        [
            "information_multiplier",
            "method",
        ]
    )
    .reset_index(
        drop=True
    )
)


print()
print(
    "Stratified screening summary"
)

display(
    STRATIFIED_SUMMARY.round(
        4
    )
)


# ============================================================
# 9. DIRECT COMPARISON TABLE
# ============================================================

PLUGIN_SUMMARY = (

    STRATIFIED_SUMMARY.loc[
        STRATIFIED_SUMMARY[
            "method"
        ]
        == "Plug-in gate"
    ]

    .set_index(
        "information_multiplier"
    )
)


ROBUST_SUMMARY = (

    STRATIFIED_SUMMARY.loc[
        STRATIFIED_SUMMARY[
            "method"
        ]
        == "Robust simultaneous gate"
    ]

    .set_index(
        "information_multiplier"
    )
)


DIRECT_COMPARISON = pd.DataFrame(
    {

        "relative_information":
            INFORMATION_MULTIPLIERS_STRATIFIED,


        "plugin_false_elimination":
            PLUGIN_SUMMARY.loc[
                INFORMATION_MULTIPLIERS_STRATIFIED,
                "false_elimination",
            ].to_numpy(),


        "robust_false_elimination":
            ROBUST_SUMMARY.loc[
                INFORMATION_MULTIPLIERS_STRATIFIED,
                "false_elimination",
            ].to_numpy(),


        "plugin_near_boundary_power":
            PLUGIN_SUMMARY.loc[
                INFORMATION_MULTIPLIERS_STRATIFIED,
                "near_boundary_power",
            ].to_numpy(),


        "robust_near_boundary_power":
            ROBUST_SUMMARY.loc[
                INFORMATION_MULTIPLIERS_STRATIFIED,
                "near_boundary_power",
            ].to_numpy(),


        "plugin_separated_power":
            PLUGIN_SUMMARY.loc[
                INFORMATION_MULTIPLIERS_STRATIFIED,
                "separated_power",
            ].to_numpy(),


        "robust_separated_power":
            ROBUST_SUMMARY.loc[
                INFORMATION_MULTIPLIERS_STRATIFIED,
                "separated_power",
            ].to_numpy(),


        "plugin_oracle_recovery":
            PLUGIN_SUMMARY.loc[
                INFORMATION_MULTIPLIERS_STRATIFIED,
                "oracle_set_recovery",
            ].to_numpy(),


        "robust_oracle_recovery":
            ROBUST_SUMMARY.loc[
                INFORMATION_MULTIPLIERS_STRATIFIED,
                "oracle_set_recovery",
            ].to_numpy(),
    }
)


DIRECT_COMPARISON[
    "false_elimination_difference_pp"
] = (

    100.0

    * (
        DIRECT_COMPARISON[
            "plugin_false_elimination"
        ]

        - DIRECT_COMPARISON[
            "robust_false_elimination"
        ]
    )
)


print()
print(
    "Direct comparison"
)

display(
    DIRECT_COMPARISON.round(
        4
    )
)


# ============================================================
# 10. FIGURE 1
#
# POLICY CURVES AND FALSE-ELIMINATION CONTROL
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        13,
        5,
    ),
)


# ------------------------------------------------------------
# Panel A
# Risk curves
# ------------------------------------------------------------

NEAR_COLORS = [
    "tab:orange",
    "tab:green",
    "tab:red",
    "tab:purple",
]


near_color_map = {

    policy:
        NEAR_COLORS[index]

    for index, policy
    in enumerate(
        sorted(
            NEAR_BOUNDARY_POLICIES
        )
    )
}


for policy_name, result in (
    TRUE_POLICY_RESULTS.items()
):

    policy_class = (
        POLICIES[
            policy_name
        ][
            "decision_class"
        ]
    )


    if policy_name == BASELINE_POLICY:

        color = (
            "tab:blue"
        )

        linestyle = (
            "-"
        )

        linewidth = 3.0

        alpha = 1.0


    elif policy_class == "Near-boundary":

        color = (
            near_color_map[
                policy_name
            ]
        )

        linestyle = (
            "-"
        )

        linewidth = 2.0

        alpha = 1.0


    else:

        color = (
            "0.55"
        )

        linestyle = (
            "--"
        )

        linewidth = 1.6

        alpha = 0.85


    axes[0].plot(

        THRESHOLDS,

        result[
            "risk_vector"
        ],

        marker="o",

        linewidth=
            linewidth,

        linestyle=
            linestyle,

        color=
            color,

        alpha=
            alpha,

        label=
            policy_name,
    )


axes[0].set_xlabel(
    "Final-size threshold"
)

axes[0].set_ylabel(
    r"$\Pr(N^I>c)$"
)

axes[0].set_ylim(
    -0.01,
    0.60,
)

axes[0].grid(
    alpha=0.20
)

axes[0].legend(
    frameon=False,
    fontsize=6.8,
)


# ------------------------------------------------------------
# Panel B
# Familywise false elimination
# ------------------------------------------------------------

for method_name, group in (
    STRATIFIED_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    axes[1].errorbar(

        group[
            "information_multiplier"
        ],

        group[
            "false_elimination"
        ],

        yerr=
            1.96
            * group[
                "false_elimination_mcse"
            ],

        marker="o",

        linewidth=2,

        capsize=4,

        label=
            method_name,
    )


axes[1].axhline(
    ALPHA,
    color="black",
    linestyle="--",
    linewidth=1.2,
    label="0.05 reference",
)


axes[1].set_xlabel(
    "Relative information"
)

axes[1].set_ylabel(
    "False-elimination probability"
)

axes[1].set_ylim(
    -0.01,
    0.50,
)

axes[1].grid(
    alpha=0.20
)

axes[1].legend(
    frameon=False
)


fig.tight_layout()


fig.savefig(
    STRATIFIED_OUTPUT_DIR
    / "experiment3_stratified_false_elimination.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 11. FIGURE 2
#
# THE IMPORTANT NEW FIGURE
#
# SEPARATES DIFFICULT AND CLEAR DOMINANCE COMPARISONS
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(
        15,
        4.7,
    ),
)


for method_name, group in (
    STRATIFIED_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    # --------------------------------------------------------
    # Near-boundary power
    # --------------------------------------------------------

    axes[0].errorbar(

        group[
            "information_multiplier"
        ],

        group[
            "near_boundary_power"
        ],

        yerr=
            1.96
            * group[
                "near_boundary_power_mcse"
            ],

        marker="o",

        linewidth=2,

        capsize=3,

        label=
            method_name,
    )


    # --------------------------------------------------------
    # Separated-alternative power
    # --------------------------------------------------------

    axes[1].errorbar(

        group[
            "information_multiplier"
        ],

        group[
            "separated_power"
        ],

        yerr=
            1.96
            * group[
                "separated_power_mcse"
            ],

        marker="o",

        linewidth=2,

        capsize=3,

        label=
            method_name,
    )


    # --------------------------------------------------------
    # Oracle-set recovery
    # --------------------------------------------------------

    axes[2].errorbar(

        group[
            "information_multiplier"
        ],

        group[
            "oracle_set_recovery"
        ],

        yerr=
            1.96
            * group[
                "oracle_set_recovery_mcse"
            ],

        marker="o",

        linewidth=2,

        capsize=3,

        label=
            method_name,
    )


axes[0].set_ylabel(
    "Near-boundary screening power"
)


axes[1].set_ylabel(
    "Separated-alternative screening power"
)


axes[2].set_ylabel(
    "Oracle-set recovery probability"
)


for axis in axes:

    axis.set_xlabel(
        "Relative information"
    )

    axis.set_ylim(
        -0.01,
        1.01,
    )

    axis.grid(
        alpha=0.20
    )

    axis.legend(
        frameon=False,
        fontsize=8,
    )


fig.tight_layout()


fig.savefig(
    STRATIFIED_OUTPUT_DIR
    / "experiment3_stratified_power.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 12. FIGURE 3
#
# PROTECTION VERSUS DISCRIMINATION
#
# This is potentially the best main-paper figure.
# ============================================================

fig, ax = plt.subplots(
    figsize=(
        7.5,
        5.5,
    )
)


for method_name, group in (
    STRATIFIED_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    ax.plot(

        group[
            "false_elimination"
        ],

        group[
            "separated_power"
        ],

        marker="o",

        linewidth=2.2,

        label=
            method_name,
    )


    for _, row in group.iterrows():

        ax.annotate(

            f"{row['information_multiplier']:g}",

            (
                row[
                    "false_elimination"
                ],

                row[
                    "separated_power"
                ],
            ),

            xytext=(
                5,
                5,
            ),

            textcoords=
                "offset points",

            fontsize=8,
        )


ax.axvline(
    ALPHA,
    color="black",
    linestyle="--",
    linewidth=1.1,
)


ax.set_xlabel(
    "False-elimination probability"
)

ax.set_ylabel(
    "Separated-alternative screening power"
)

ax.set_xlim(
    -0.01,
    max(
        0.35,
        1.05
        * STRATIFIED_SUMMARY[
            "false_elimination"
        ].max(),
    ),
)

ax.set_ylim(
    -0.01,
    1.01,
)

ax.grid(
    alpha=0.20
)

ax.legend(
    frameon=False
)


fig.tight_layout()


fig.savefig(
    STRATIFIED_OUTPUT_DIR
    / "experiment3_protection_discrimination_tradeoff.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 13. SAVE RESULTS
# ============================================================

STRATIFIED_POLICY_DESIGN.to_csv(
    STRATIFIED_OUTPUT_DIR
    / "experiment3_stratified_policy_design.csv",
    index=False,
)


ORACLE_PAIRWISE.to_csv(
    STRATIFIED_OUTPUT_DIR
    / "experiment3_stratified_oracle_pairwise.csv",
    index=False,
)


STRATIFIED_RESULTS.to_csv(
    STRATIFIED_OUTPUT_DIR
    / "experiment3_stratified_replication_results.csv",
    index=False,
)


STRATIFIED_SUMMARY.to_csv(
    STRATIFIED_OUTPUT_DIR
    / "experiment3_stratified_summary.csv",
    index=False,
)


DIRECT_COMPARISON.to_csv(
    STRATIFIED_OUTPUT_DIR
    / "experiment3_stratified_direct_comparison.csv",
    index=False,
)


print()
print(
    "Experiment 3B completed"
)

print(
    "Results saved in",
    STRATIFIED_OUTPUT_DIR,
)
```
