```python
# ============================================================
# EXPERIMENT 3
#
# UNCERTAINTY-AWARE INTERVENTION SCREENING
#
# FINAL VERSION CONDITIONAL ON EXPERIMENT 1
#
# Fixed from Experiment 1
#
# N_eff = 175
# S(0)  = 114
# I(0)  = 1
# R(0)  = 60
#
# b_hat
# gamma_hat
# phi_hat
# Sigma_eta
# Hessian
#
# No data loading
# No likelihood fitting
# No Hessian estimation
#
# The intervention gate is based ONLY on the strict
# final-size tail-risk curve
#
# rho_a(c) = P_a(N^I > c)
#
# The final-size mean and extinction-time mean are reported
# descriptively but are NOT used in the dominance rule.
#
# One autonomous Google Colab cell
# ============================================================


from __future__ import annotations

import time
from pathlib import Path
from itertools import combinations

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from IPython.display import display


TOTAL_START = time.perf_counter()


# ============================================================
# 1. FIXED INPUTS FROM EXPERIMENT 1
# ============================================================

N = 175
S0 = 114
I0 = 1
R0_INITIAL = 60

assert N == S0 + I0 + R0_INITIAL


# Exact conditional parameter estimates

B_HAT = 0.7650254997116541
GAMMA_HAT = 0.23768509187018172
PHI_HAT = 2.4906108661339097

EXACT_PENALIZED_NLL = 45.351238703709555


# Parameter ordering
#
# eta = (
#     log b,
#     log gamma,
#     log phi
# )

SIGMA_ETA = np.array(
    [
        [
            0.01078625,
            0.01032878,
            -0.01802093,
        ],
        [
            0.01032878,
            0.01670211,
            -0.03299136,
        ],
        [
            -0.01802093,
            -0.03299136,
            0.55918411,
        ],
    ],
    dtype=float,
)


# Hessian implied by the fixed covariance matrix

HESSIAN_ETA = np.linalg.inv(
    SIGMA_ETA
)


# Only b and gamma determine the stochastic
# final-size and extinction descriptors.

SIGMA_BG = SIGMA_ETA[
    np.ix_(
        [0, 1],
        [0, 1],
    )
]


ETA_BG_HAT = np.log(
    [
        B_HAT,
        GAMMA_HAT,
    ]
)


# ============================================================
# 2. EXPERIMENT 1 REFERENCE VALUES
# ============================================================

EXP1_FINAL_SIZE_MEAN = 47.624525

EXP1_EXTINCTION_MEAN = 23.658440

EXP1_TAIL_RISKS = np.array(
    [
        0.502185,
        0.492970,
        0.454210,
        0.318310,
        0.045407,
    ],
    dtype=float,
)


# Same prespecified thresholds as Experiment 1

THRESHOLD_FRACTIONS = np.array(
    [
        0.50,
        0.60,
        0.70,
        0.80,
        0.90,
    ],
    dtype=float,
)


THRESHOLDS = np.round(
    THRESHOLD_FRACTIONS
    * S0
).astype(int)


assert np.array_equal(
    THRESHOLDS,
    np.array(
        [
            57,
            68,
            80,
            91,
            103,
        ]
    ),
)


# ============================================================
# 3. SIMULATION SETTINGS
# ============================================================

ALPHA = 0.05

BASE_SEED = 20260807


# Large number for the fitted Cameroon application

N_APPLICATION_DRAWS = 5000


# The manuscript intervention-screening experiment uses
# Monte Carlo replication and one-sided max-t resampling.

N_REPLICATIONS = 200

N_BOOTSTRAP_DRAWS = 299


# Relative amounts of parameter information.
#
# Covariance is scaled as
#
# Sigma / information_multiplier
#
# Thus values above one represent more information.

INFORMATION_MULTIPLIERS = np.array(
    [
        0.50,
        1.00,
        2.00,
        4.00,
    ],
    dtype=float,
)


OUTPUT_DIR = Path(
    "/content/experiment3_fixed_from_experiment1"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


print(
    "Fixed Experiment 1 inputs"
)

print(
    {
        "N_eff": N,
        "S0": S0,
        "I0": I0,
        "R0": R0_INITIAL,
        "b_hat": B_HAT,
        "gamma_hat": GAMMA_HAT,
        "phi_hat": PHI_HAT,
        "exact_penalized_nll":
            EXACT_PENALIZED_NLL,
    }
)


print()

print(
    "Fixed covariance of log b and log gamma"
)

print(
    SIGMA_BG
)


print()

print(
    "Fixed Hessian eigenvalues"
)

print(
    np.linalg.eigvalsh(
        HESSIAN_ETA
    )
)


# ============================================================
# 4. INTERVENTION SCENARIOS
# ============================================================

# These are hypothetical parameter transformations.
#
# They do not alter N, S(0), I(0), or R(0).
#
# Multipliers act on the fitted b and gamma.

POLICIES = {

    "No additional intervention": {

        "tier":
            "Reference",

        "b_multiplier":
            1.00,

        "gamma_multiplier":
            1.00,
    },


    "Moderate contact reduction": {

        "tier":
            "Moderate",

        "b_multiplier":
            0.80,

        "gamma_multiplier":
            1.00,
    },


    "Moderate rapid isolation": {

        "tier":
            "Moderate",

        "b_multiplier":
            1.00,

        "gamma_multiplier":
            1.25,
    },


    "Moderate balanced response": {

        "tier":
            "Moderate",

        "b_multiplier":
            0.90,

        "gamma_multiplier":
            1.12,
    },


    "Strong contact reduction": {

        "tier":
            "Strong",

        "b_multiplier":
            0.65,

        "gamma_multiplier":
            1.00,
    },


    "Strong rapid isolation": {

        "tier":
            "Strong",

        "b_multiplier":
            1.00,

        "gamma_multiplier":
            1.55,
    },


    "Strong balanced response": {

        "tier":
            "Strong",

        "b_multiplier":
            0.80,

        "gamma_multiplier":
            1.23,
    },
}


BASELINE_POLICY = (
    "No additional intervention"
)


POLICY_TIERS = {

    "Moderate": [

        name

        for name, specification
        in POLICIES.items()

        if specification[
            "tier"
        ] == "Moderate"
    ],


    "Strong": [

        name

        for name, specification
        in POLICIES.items()

        if specification[
            "tier"
        ] == "Strong"
    ],
}


# ============================================================
# 5. EXACT FINAL-SIZE PMF AND RECURSIVE GRADIENT
# ============================================================

def final_size_pmf_gradient(
    b: float,
    gamma: float,
) -> tuple[
    np.ndarray,
    np.ndarray,
]:

    """
    Exact final-size PMF and its recursive derivatives
    with respect to log b and log gamma.

    pmf[n]
        P(N^I = n)

    gradient[0,n]
        derivative with respect to log b

    gradient[1,n]
        derivative with respect to log gamma
    """


    probability = np.zeros(
        I0 + 1,
        dtype=float,
    )

    probability[
        I0
    ] = 1.0


    gradient = np.zeros(
        (
            2,
            I0 + 1,
        ),
        dtype=float,
    )


    pmf = np.zeros(
        S0 + 1,
        dtype=float,
    )


    pmf_gradient = np.zeros(
        (
            2,
            S0 + 1,
        ),
        dtype=float,
    )


    for susceptible in range(
        S0,
        -1,
        -1,
    ):

        maximum_infectious = (
            I0
            + S0
            - susceptible
        )


        required_length = (
            maximum_infectious
            + 1
        )


        if len(
            probability
        ) < required_length:

            padding = (
                required_length
                - len(
                    probability
                )
            )


            probability = np.pad(
                probability,
                (
                    0,
                    padding,
                ),
            )


            gradient = np.pad(
                gradient,
                (
                    (0, 0),
                    (0, padding),
                ),
            )


        if susceptible > 0:

            infection_component = (
                b
                * susceptible
                / N
            )


            denominator = (
                infection_component
                + gamma
            )


            infection_probability = (
                infection_component
                / denominator
            )


            recovery_probability = (
                gamma
                / denominator
            )


            derivative_infection = np.array(
                [
                    infection_probability
                    * recovery_probability,

                    -infection_probability
                    * recovery_probability,
                ],
                dtype=float,
            )


            derivative_recovery = (
                -derivative_infection
            )


            next_probability = np.zeros(
                maximum_infectious
                + 2,
                dtype=float,
            )


            next_gradient = np.zeros(
                (
                    2,
                    maximum_infectious
                    + 2,
                ),
                dtype=float,
            )


        else:

            infection_probability = 0.0
            recovery_probability = 1.0


            derivative_infection = np.zeros(
                2,
                dtype=float,
            )


            derivative_recovery = np.zeros(
                2,
                dtype=float,
            )


            next_probability = None
            next_gradient = None


        for infectious in range(
            maximum_infectious,
            0,
            -1,
        ):

            mass = probability[
                infectious
            ]


            mass_gradient = gradient[
                :,
                infectious,
            ].copy()


            if (
                abs(
                    mass
                ) < 1e-300

                and np.all(
                    np.abs(
                        mass_gradient
                    )
                    < 1e-300
                )
            ):

                continue


            # Recovery transition

            probability[
                infectious - 1
            ] += (
                mass
                * recovery_probability
            )


            gradient[
                :,
                infectious - 1,
            ] += (

                mass_gradient
                * recovery_probability

                + mass
                * derivative_recovery
            )


            # Infection transition

            if susceptible > 0:

                next_probability[
                    infectious + 1
                ] += (
                    mass
                    * infection_probability
                )


                next_gradient[
                    :,
                    infectious + 1,
                ] += (

                    mass_gradient
                    * infection_probability

                    + mass
                    * derivative_infection
                )


        infections_so_far = (
            S0
            - susceptible
        )


        pmf[
            infections_so_far
        ] = (
            probability[0]
        )


        pmf_gradient[
            :,
            infections_so_far,
        ] = gradient[
            :,
            0,
        ]


        if susceptible > 0:

            probability = (
                next_probability
            )

            gradient = (
                next_gradient
            )


    normalizer = float(
        pmf.sum()
    )


    normalizer_gradient = (
        pmf_gradient.sum(
            axis=1
        )
    )


    normalized_pmf = (
        pmf
        / normalizer
    )


    normalized_gradient = (

        pmf_gradient
        * normalizer

        - pmf[
            None,
            :
        ]
        * normalizer_gradient[
            :,
            None,
        ]

    ) / (
        normalizer**2
    )


    return (
        normalized_pmf,
        normalized_gradient,
    )


# ============================================================
# 6. STRICT TAIL PROBABILITIES
#
# rho(c) = P(N^I > c)
# ============================================================

def strict_tail_and_gradient(
    pmf: np.ndarray,
    pmf_gradient: np.ndarray,
) -> tuple[
    np.ndarray,
    np.ndarray,
]:

    nonstrict_tail = np.cumsum(
        pmf[
            ::-1
        ]
    )[
        ::-1
    ]


    nonstrict_gradient = np.cumsum(
        pmf_gradient[
            :,
            ::-1,
        ],
        axis=1,
    )[
        :,
        ::-1,
    ]


    strict_tail = (
        nonstrict_tail
        - pmf
    )


    strict_gradient = (
        nonstrict_gradient
        - pmf_gradient
    )


    return (
        strict_tail,
        strict_gradient,
    )


# ============================================================
# 7. EXPECTED EXTINCTION TIME AND GRADIENT
# ============================================================

def extinction_mean_gradient(
    b: float,
    gamma: float,
) -> tuple[
    float,
    np.ndarray,
]:

    previous_mean = None
    previous_gradient = None


    for susceptible in range(
        0,
        S0 + 1,
    ):

        maximum_infectious = (
            I0
            + S0
            - susceptible
        )


        mean = np.zeros(
            maximum_infectious
            + 1,
            dtype=float,
        )


        gradient = np.zeros(
            (
                2,
                maximum_infectious
                + 1,
            ),
            dtype=float,
        )


        infection_component = (
            b
            * susceptible
            / N
        )


        denominator = (
            infection_component
            + gamma
        )


        infection_probability = (
            infection_component
            / denominator

            if susceptible > 0

            else 0.0
        )


        recovery_probability = (
            1.0
            - infection_probability
        )


        derivative_infection = np.array(
            [
                infection_probability
                * recovery_probability,

                -infection_probability
                * recovery_probability,
            ],
            dtype=float,
        )


        for infectious in range(
            1,
            maximum_infectious + 1,
        ):

            total_rate = (
                infectious
                * denominator
            )


            holding_mean = (
                1.0
                / total_rate
            )


            holding_gradient = (

                -holding_mean

                * np.array(
                    [
                        infection_probability,
                        recovery_probability,
                    ],
                    dtype=float,
                )
            )


            recovery_mean = mean[
                infectious - 1
            ]


            recovery_gradient = gradient[
                :,
                infectious - 1,
            ]


            if susceptible > 0:

                infection_mean = (
                    previous_mean[
                        infectious + 1
                    ]
                )


                infection_gradient = (
                    previous_gradient[
                        :,
                        infectious + 1,
                    ]
                )


            else:

                infection_mean = 0.0

                infection_gradient = (
                    np.zeros(
                        2,
                        dtype=float,
                    )
                )


            next_mean = (

                infection_probability
                * infection_mean

                + recovery_probability
                * recovery_mean
            )


            next_gradient = (

                derivative_infection
                * (
                    infection_mean
                    - recovery_mean
                )

                + infection_probability
                * infection_gradient

                + recovery_probability
                * recovery_gradient
            )


            mean[
                infectious
            ] = (

                holding_mean
                + next_mean
            )


            gradient[
                :,
                infectious,
            ] = (

                holding_gradient
                + next_gradient
            )


        previous_mean = mean
        previous_gradient = gradient


    return (
        float(
            mean[
                I0
            ]
        ),

        gradient[
            :,
            I0,
        ].copy(),
    )


# ============================================================
# 8. POLICY EVALUATION
# ============================================================

def evaluate_policy(
    eta_bg: np.ndarray,
    policy_name: str,
) -> dict:

    specification = (
        POLICIES[
            policy_name
        ]
    )


    baseline_b = float(
        np.exp(
            eta_bg[0]
        )
    )


    baseline_gamma = float(
        np.exp(
            eta_bg[1]
        )
    )


    policy_b = (

        baseline_b

        * specification[
            "b_multiplier"
        ]
    )


    policy_gamma = (

        baseline_gamma

        * specification[
            "gamma_multiplier"
        ]
    )


    (
        pmf,
        pmf_gradient,
    ) = final_size_pmf_gradient(
        b=policy_b,
        gamma=policy_gamma,
    )


    (
        strict_tail,
        strict_tail_gradient,
    ) = strict_tail_and_gradient(
        pmf=pmf,
        pmf_gradient=
            pmf_gradient,
    )


    risk_vector = (
        strict_tail[
            THRESHOLDS
        ]
    )


    risk_gradient = (
        strict_tail_gradient[
            :,
            THRESHOLDS,
        ].T
    )


    final_size_values = np.arange(
        len(
            pmf
        ),
        dtype=float,
    )


    expected_final_size = float(
        np.sum(
            final_size_values
            * pmf
        )
    )


    final_size_gradient = np.sum(
        final_size_values[
            None,
            :
        ]
        * pmf_gradient,
        axis=1,
    )


    (
        expected_extinction_time,
        extinction_gradient,
    ) = extinction_mean_gradient(
        b=policy_b,
        gamma=policy_gamma,
    )


    initial_effective_ratio = (
        policy_b
        * S0
        / (
            policy_gamma
            * N
        )
    )


    return {

        "policy":
            policy_name,

        "tier":
            specification[
                "tier"
            ],

        "b":
            policy_b,

        "gamma":
            policy_gamma,

        "initial_effective_ratio":
            initial_effective_ratio,

        "risk_vector":
            risk_vector,

        "risk_gradient":
            risk_gradient,

        "expected_final_size":
            expected_final_size,

        "final_size_gradient":
            final_size_gradient,

        "expected_extinction_time":
            expected_extinction_time,

        "extinction_gradient":
            extinction_gradient,

        "pmf":
            pmf,

        "strict_tail":
            strict_tail,
    }


def evaluate_all_policies(
    eta_bg: np.ndarray,
) -> dict:

    return {

        policy_name:
            evaluate_policy(
                eta_bg=
                    eta_bg,
                policy_name=
                    policy_name,
            )

        for policy_name
        in POLICIES
    }


FITTED_POLICIES = (
    evaluate_all_policies(
        ETA_BG_HAT
    )
)


# ============================================================
# 9. INTERNAL AGREEMENT WITH EXPERIMENT 1
# ============================================================

BASELINE_RESULT = (
    FITTED_POLICIES[
        BASELINE_POLICY
    ]
)


print()
print(
    "Experiment 1 consistency checks"
)


print(
    {
        "max_tail_risk_difference":
            float(
                np.max(
                    np.abs(
                        BASELINE_RESULT[
                            "risk_vector"
                        ]
                        - EXP1_TAIL_RISKS
                    )
                )
            ),

        "final_size_difference":
            float(
                BASELINE_RESULT[
                    "expected_final_size"
                ]
                - EXP1_FINAL_SIZE_MEAN
            ),

        "extinction_difference":
            float(
                BASELINE_RESULT[
                    "expected_extinction_time"
                ]
                - EXP1_EXTINCTION_MEAN
            ),
    }
)


if np.max(
    np.abs(
        BASELINE_RESULT[
            "risk_vector"
        ]
        - EXP1_TAIL_RISKS
    )
) > 5e-4:

    raise RuntimeError(
        "Baseline tail risks do not reproduce Experiment 1."
    )


if abs(
    BASELINE_RESULT[
        "expected_final_size"
    ]
    - EXP1_FINAL_SIZE_MEAN
) > 5e-4:

    raise RuntimeError(
        "Expected final size does not reproduce Experiment 1."
    )


if abs(
    BASELINE_RESULT[
        "expected_extinction_time"
    ]
    - EXP1_EXTINCTION_MEAN
) > 5e-4:

    raise RuntimeError(
        "Expected extinction time does not reproduce Experiment 1."
    )


# ============================================================
# 10. POLICY DESCRIPTOR TABLE
# ============================================================

APPLICATION_ROWS = []


for policy_name, result in (
    FITTED_POLICIES.items()
):

    row = {

        "policy":
            policy_name,

        "tier":
            result[
                "tier"
            ],

        "b":
            result[
                "b"
            ],

        "gamma":
            result[
                "gamma"
            ],

        "initial_effective_ratio":
            result[
                "initial_effective_ratio"
            ],

        "expected_final_size":
            result[
                "expected_final_size"
            ],

        "expected_extinction_time":
            result[
                "expected_extinction_time"
            ],
    }


    for threshold_index, threshold in enumerate(
        THRESHOLDS
    ):

        row[
            f"Pr_NI_gt_{threshold}"
        ] = (
            result[
                "risk_vector"
            ][
                threshold_index
            ]
        )


    APPLICATION_ROWS.append(
        row
    )


APPLICATION_SUMMARY = pd.DataFrame(
    APPLICATION_ROWS
)


print()
print(
    "Policy descriptors"
)

display(
    APPLICATION_SUMMARY.round(
        5
    )
)


# ============================================================
# 11. PLUG-IN DOMINANCE
#
# Dominance uses ONLY the risk curve.
# ============================================================

def dominates_vector(
    first: np.ndarray,
    second: np.ndarray,
    tolerance: float = 1e-10,
) -> bool:

    return bool(

        np.all(
            first
            <= second
            + tolerance
        )

        and np.any(
            first
            < second
            - tolerance
        )
    )


def plugin_gate(
    policy_names: list[str],
    policy_results: dict,
) -> tuple[
    set[str],
    pd.DataFrame,
]:

    eliminated = set()

    rows = []


    for first in policy_names:

        for second in policy_names:

            if first == second:
                continue


            first_curve = (
                policy_results[
                    first
                ][
                    "risk_vector"
                ]
            )


            second_curve = (
                policy_results[
                    second
                ][
                    "risk_vector"
                ]
            )


            contrast = (
                first_curve
                - second_curve
            )


            dominance = (
                dominates_vector(
                    first_curve,
                    second_curve,
                )
            )


            if dominance:

                eliminated.add(
                    second
                )


            rows.append(
                {
                    "potential_dominator":
                        first,

                    "potentially_dominated":
                        second,

                    "maximum_contrast":
                        float(
                            np.max(
                                contrast
                            )
                        ),

                    "minimum_contrast":
                        float(
                            np.min(
                                contrast
                            )
                        ),

                    "plugin_dominance":
                        dominance,
                }
            )


    retained = (
        set(
            policy_names
        )
        - eliminated
    )


    return (
        retained,
        pd.DataFrame(
            rows
        ),
    )


# ============================================================
# 12. GLOBAL ONE-SIDED MAX-T ROBUST GATE
#
# For each ordered intervention pair
#
# Delta_lk(c)
# =
# rho_l(c) - rho_k(c)
#
# The upper simultaneous band is
#
# Delta_hat + q * SE
#
# Intervention k is removed only when another intervention l
# has upper band <= 0 at every threshold and strict
# improvement at at least one threshold.
# ============================================================

def robust_gate(
    policy_names: list[str],
    policy_results: dict,
    covariance: np.ndarray,
    rng: np.random.Generator,
    n_draws: int,
) -> tuple[
    set[str],
    pd.DataFrame,
    float,
]:

    ordered_pairs = [

        (
            first,
            second,
        )

        for first in policy_names

        for second in policy_names

        if first != second
    ]


    gaussian_draws = (
        rng.multivariate_normal(
            mean=np.zeros(
                2
            ),
            cov=covariance,
            size=n_draws,
        )
    )


    pair_information = []

    standardized_collection = []


    for first, second in (
        ordered_pairs
    ):

        first_result = (
            policy_results[
                first
            ]
        )


        second_result = (
            policy_results[
                second
            ]
        )


        contrast = (

            first_result[
                "risk_vector"
            ]

            - second_result[
                "risk_vector"
            ]
        )


        contrast_gradient = (

            first_result[
                "risk_gradient"
            ]

            - second_result[
                "risk_gradient"
            ]
        )


        contrast_variance = np.maximum(
            np.einsum(
                "ki,ij,kj->k",
                contrast_gradient,
                covariance,
                contrast_gradient,
            ),
            0.0,
        )


        standard_errors = np.sqrt(
            contrast_variance
        )


        perturbations = (

            gaussian_draws

            @ contrast_gradient.T
        )


        active = (
            standard_errors
            > 1e-12
        )


        standardized = np.zeros_like(
            perturbations
        )


        standardized[
            :,
            active,
        ] = (

            perturbations[
                :,
                active,
            ]

            / standard_errors[
                active
            ]
        )


        # Only non-degenerate components contribute
        # to the max statistic.

        if np.any(
            active
        ):

            standardized_collection.append(
                standardized[
                    :,
                    active,
                ]
            )


        pair_information.append(
            {
                "first":
                    first,

                "second":
                    second,

                "contrast":
                    contrast,

                "contrast_gradient":
                    contrast_gradient,

                "standard_errors":
                    standard_errors,

                "active":
                    active,
            }
        )


    if not standardized_collection:

        critical_value = 0.0

    else:

        all_standardized = np.concatenate(
            standardized_collection,
            axis=1,
        )


        max_statistics = np.max(
            all_standardized,
            axis=1,
        )


        critical_value = float(
            np.quantile(
                max_statistics,
                1.0 - ALPHA,
            )
        )


    eliminated = set()

    rows = []


    for information in (
        pair_information
    ):

        first = information[
            "first"
        ]

        second = information[
            "second"
        ]

        contrast = information[
            "contrast"
        ]

        standard_errors = information[
            "standard_errors"
        ]


        upper_band = (

            contrast

            + critical_value
            * standard_errors
        )


        dominance_supported = bool(

            np.all(
                upper_band
                <= 1e-10
            )

            and np.any(
                upper_band
                < -1e-10
            )
        )


        if dominance_supported:

            eliminated.add(
                second
            )


        rows.append(
            {
                "potential_dominator":
                    first,

                "potentially_dominated":
                    second,

                "maximum_estimated_contrast":
                    float(
                        np.max(
                            contrast
                        )
                    ),

                "maximum_upper_band":
                    float(
                        np.max(
                            upper_band
                        )
                    ),

                "minimum_upper_band":
                    float(
                        np.min(
                            upper_band
                        )
                    ),

                "simultaneous_dominance":
                    dominance_supported,
            }
        )


    retained = (
        set(
            policy_names
        )
        - eliminated
    )


    return (
        retained,
        pd.DataFrame(
            rows
        ),
        critical_value,
    )


# ============================================================
# 13. AMBIGUITY RATE
# ============================================================

def ambiguity_rate(
    policy_names: list[str],
    pairwise_table: pd.DataFrame,
    dominance_column: str,
) -> float:

    ambiguous = []


    for first, second in combinations(
        policy_names,
        2,
    ):

        first_to_second = pairwise_table.loc[
            (
                pairwise_table[
                    "potential_dominator"
                ] == first
            )
            & (
                pairwise_table[
                    "potentially_dominated"
                ] == second
            ),
            dominance_column,
        ]


        second_to_first = pairwise_table.loc[
            (
                pairwise_table[
                    "potential_dominator"
                ] == second
            )
            & (
                pairwise_table[
                    "potentially_dominated"
                ] == first
            ),
            dominance_column,
        ]


        dominates_either_way = (

            bool(
                first_to_second.iloc[0]
            )

            or bool(
                second_to_first.iloc[0]
            )
        )


        ambiguous.append(
            not dominates_either_way
        )


    if not ambiguous:

        return np.nan


    return float(
        np.mean(
            ambiguous
        )
    )


# ============================================================
# 14. CAMEROON APPLICATION GATE
# ============================================================

APPLICATION_RNG = np.random.default_rng(
    BASE_SEED
)


APPLICATION_GATE_ROWS = []

PAIRWISE_APPLICATION_TABLES = []


for tier_name, policy_names in (
    POLICY_TIERS.items()
):

    (
        plugin_retained,
        plugin_pairwise,
    ) = plugin_gate(
        policy_names=
            policy_names,
        policy_results=
            FITTED_POLICIES,
    )


    (
        robust_retained,
        robust_pairwise,
        critical_value,
    ) = robust_gate(
        policy_names=
            policy_names,
        policy_results=
            FITTED_POLICIES,
        covariance=
            SIGMA_BG,
        rng=
            APPLICATION_RNG,
        n_draws=
            N_APPLICATION_DRAWS,
    )


    plugin_ambiguity = (
        ambiguity_rate(
            policy_names=
                policy_names,
            pairwise_table=
                plugin_pairwise,
            dominance_column=
                "plugin_dominance",
        )
    )


    robust_ambiguity = (
        ambiguity_rate(
            policy_names=
                policy_names,
            pairwise_table=
                robust_pairwise,
            dominance_column=
                "simultaneous_dominance",
        )
    )


    plugin_pairwise[
        "tier"
    ] = tier_name


    robust_pairwise[
        "tier"
    ] = tier_name


    merged_pairwise = (
        plugin_pairwise.merge(
            robust_pairwise,
            on=[
                "tier",
                "potential_dominator",
                "potentially_dominated",
            ],
            how="outer",
        )
    )


    PAIRWISE_APPLICATION_TABLES.append(
        merged_pairwise
    )


    APPLICATION_GATE_ROWS.append(
        {
            "tier":
                tier_name,

            "plugin_retained":
                ", ".join(
                    sorted(
                        plugin_retained
                    )
                ),

            "robust_retained":
                ", ".join(
                    sorted(
                        robust_retained
                    )
                ),

            "plugin_retained_count":
                len(
                    plugin_retained
                ),

            "robust_retained_count":
                len(
                    robust_retained
                ),

            "plugin_ambiguity_rate":
                plugin_ambiguity,

            "robust_ambiguity_rate":
                robust_ambiguity,

            "global_one_sided_critical_value":
                critical_value,
        }
    )


APPLICATION_GATE_SUMMARY = pd.DataFrame(
    APPLICATION_GATE_ROWS
)


PAIRWISE_GATE_RESULTS = pd.concat(
    PAIRWISE_APPLICATION_TABLES,
    ignore_index=True,
)


print()
print(
    "Application intervention gate"
)

display(
    APPLICATION_GATE_SUMMARY.round(
        5
    )
)


print()
print(
    "Pairwise intervention comparisons"
)

display(
    PAIRWISE_GATE_RESULTS.round(
        5
    )
)


# ============================================================
# 15. SIMULTANEOUS CONTRASTS AGAINST BASELINE
# ============================================================

NONBASELINE_POLICIES = [

    policy_name

    for policy_name in POLICIES

    if policy_name
    != BASELINE_POLICY
]


BASELINE_RNG = np.random.default_rng(
    BASE_SEED
    + 100
)


baseline_draws = (
    BASELINE_RNG.multivariate_normal(
        mean=np.zeros(
            2
        ),
        cov=SIGMA_BG,
        size=N_APPLICATION_DRAWS,
    )
)


BASELINE_INFORMATION = []

BASELINE_STANDARDIZED = []


for policy_name in (
    NONBASELINE_POLICIES
):

    result = (
        FITTED_POLICIES[
            policy_name
        ]
    )


    baseline = (
        FITTED_POLICIES[
            BASELINE_POLICY
        ]
    )


    contrast = (

        result[
            "risk_vector"
        ]

        - baseline[
            "risk_vector"
        ]
    )


    contrast_gradient = (

        result[
            "risk_gradient"
        ]

        - baseline[
            "risk_gradient"
        ]
    )


    standard_errors = np.sqrt(
        np.maximum(
            np.einsum(
                "ki,ij,kj->k",
                contrast_gradient,
                SIGMA_BG,
                contrast_gradient,
            ),
            0.0,
        )
    )


    perturbations = (

        baseline_draws

        @ contrast_gradient.T
    )


    active = (
        standard_errors
        > 1e-12
    )


    standardized = np.zeros_like(
        perturbations
    )


    standardized[
        :,
        active,
    ] = (

        perturbations[
            :,
            active,
        ]

        / standard_errors[
            active
        ]
    )


    if np.any(
        active
    ):

        BASELINE_STANDARDIZED.append(
            standardized[
                :,
                active,
            ]
        )


    BASELINE_INFORMATION.append(
        {
            "policy":
                policy_name,

            "contrast":
                contrast,

            "standard_errors":
                standard_errors,
        }
    )


BASELINE_MAX_STATISTIC = np.max(
    np.concatenate(
        BASELINE_STANDARDIZED,
        axis=1,
    ),
    axis=1,
)


BASELINE_CRITICAL_VALUE = float(
    np.quantile(
        BASELINE_MAX_STATISTIC,
        1.0 - ALPHA,
    )
)


BASELINE_ROWS = []


for information in (
    BASELINE_INFORMATION
):

    policy_name = information[
        "policy"
    ]


    contrast = information[
        "contrast"
    ]


    standard_errors = information[
        "standard_errors"
    ]


    upper_band = (

        contrast

        + BASELINE_CRITICAL_VALUE
        * standard_errors
    )


    uniformly_beneficial = bool(

        np.all(
            upper_band
            <= 1e-10
        )

        and np.any(
            upper_band
            < -1e-10
        )
    )


    row = {

        "policy":
            policy_name,

        "uniformly_beneficial_vs_baseline":
            uniformly_beneficial,

        "maximum_estimated_contrast":
            float(
                np.max(
                    contrast
                )
            ),

        "maximum_simultaneous_upper_band":
            float(
                np.max(
                    upper_band
                )
            ),
    }


    for threshold_index, threshold in enumerate(
        THRESHOLDS
    ):

        row[
            f"contrast_{threshold}"
        ] = (
            contrast[
                threshold_index
            ]
        )


        row[
            f"upper_band_{threshold}"
        ] = (
            upper_band[
                threshold_index
            ]
        )


    BASELINE_ROWS.append(
        row
    )


BASELINE_CONTRAST_SUMMARY = pd.DataFrame(
    BASELINE_ROWS
)


print()
print(
    "Simultaneous intervention contrasts against baseline"
)

display(
    BASELINE_CONTRAST_SUMMARY.round(
        5
    )
)


# ============================================================
# 16. ORACLE NON-DOMINATED SET
#
# For the repeated-sampling experiment, the fitted
# Experiment 1 parameter vector is treated as the working
# data-generating parameter.
# ============================================================

ORACLE_NONDOMINATED = {}

ORACLE_DOMINATED = {}


for tier_name, policy_names in (
    POLICY_TIERS.items()
):

    (
        retained,
        _,
    ) = plugin_gate(
        policy_names=
            policy_names,
        policy_results=
            FITTED_POLICIES,
    )


    ORACLE_NONDOMINATED[
        tier_name
    ] = retained


    ORACLE_DOMINATED[
        tier_name
    ] = (

        set(
            policy_names
        )

        - retained
    )


print()
print(
    "Oracle non-dominated sets"
)


for tier_name in (
    POLICY_TIERS
):

    print(
        tier_name,
        sorted(
            ORACLE_NONDOMINATED[
                tier_name
            ]
        )
    )


# ============================================================
# 17. REPEATED-SAMPLING VALIDATION
#
# We study how the intervention gate behaves as the amount
# of information increases.
#
# The working estimator distribution is
#
# eta_hat ~ N(
#     eta_0,
#     Sigma_BG / information_multiplier
# )
#
# The robust rule uses a fresh one-sided global max-t
# resampling distribution in each replication.
# ============================================================

SIMULATION_RNG = np.random.default_rng(
    BASE_SEED
    + 10000
)


SCREENING_START = time.perf_counter()


SIMULATION_RECORDS = []


ORACLE_NONDOMINATED_ALL = set().union(
    *ORACLE_NONDOMINATED.values()
)


ORACLE_DOMINATED_ALL = set().union(
    *ORACLE_DOMINATED.values()
)


ORACLE_RETAINED_ALL = set(
    ORACLE_NONDOMINATED_ALL
)


for information_multiplier in (
    INFORMATION_MULTIPLIERS
):

    covariance = (

        SIGMA_BG

        / information_multiplier
    )


    print()
    print(
        "Information multiplier",
        information_multiplier,
    )


    for replication in range(
        N_REPLICATIONS
    ):

        eta_estimate = (
            SIMULATION_RNG.multivariate_normal(
                mean=
                    ETA_BG_HAT,
                cov=
                    covariance,
            )
        )


        estimated_policies = (
            evaluate_all_policies(
                eta_estimate
            )
        )


        # ----------------------------------------------------
        # Plug-in rule
        # ----------------------------------------------------

        plugin_retained_all = set()

        plugin_eliminated_all = set()

        plugin_ambiguity_values = []


        for tier_name, policy_names in (
            POLICY_TIERS.items()
        ):

            (
                retained,
                pairwise,
            ) = plugin_gate(
                policy_names=
                    policy_names,
                policy_results=
                    estimated_policies,
            )


            plugin_retained_all.update(
                retained
            )


            plugin_eliminated_all.update(

                set(
                    policy_names
                )

                - retained
            )


            plugin_ambiguity_values.append(
                ambiguity_rate(
                    policy_names=
                        policy_names,
                    pairwise_table=
                        pairwise,
                    dominance_column=
                        "plugin_dominance",
                )
            )


        # ----------------------------------------------------
        # Robust simultaneous rule
        # ----------------------------------------------------

        robust_retained_all = set()

        robust_eliminated_all = set()

        robust_ambiguity_values = []

        robust_critical_values = []


        for tier_name, policy_names in (
            POLICY_TIERS.items()
        ):

            (
                retained,
                pairwise,
                critical_value,
            ) = robust_gate(
                policy_names=
                    policy_names,
                policy_results=
                    estimated_policies,
                covariance=
                    covariance,
                rng=
                    SIMULATION_RNG,
                n_draws=
                    N_BOOTSTRAP_DRAWS,
            )


            robust_retained_all.update(
                retained
            )


            robust_eliminated_all.update(

                set(
                    policy_names
                )

                - retained
            )


            robust_ambiguity_values.append(
                ambiguity_rate(
                    policy_names=
                        policy_names,
                    pairwise_table=
                        pairwise,
                    dominance_column=
                        "simultaneous_dominance",
                )
            )


            robust_critical_values.append(
                critical_value
            )


        # ----------------------------------------------------
        # Replication-level metrics
        # ----------------------------------------------------

        for method_name, retained_set, eliminated_set, ambiguity_value in [

            (
                "Plug-in gate",
                plugin_retained_all,
                plugin_eliminated_all,
                float(
                    np.nanmean(
                        plugin_ambiguity_values
                    )
                ),
            ),

            (
                "Robust simultaneous gate",
                robust_retained_all,
                robust_eliminated_all,
                float(
                    np.nanmean(
                        robust_ambiguity_values
                    )
                ),
            ),
        ]:

            false_elimination = (

                len(
                    eliminated_set
                    & ORACLE_NONDOMINATED_ALL
                )
                > 0
            )


            if len(
                ORACLE_DOMINATED_ALL
            ) > 0:

                screening_power = (

                    len(
                        eliminated_set
                        & ORACLE_DOMINATED_ALL
                    )

                    / len(
                        ORACLE_DOMINATED_ALL
                    )
                )

            else:

                screening_power = (
                    np.nan
                )


            oracle_agreement = (

                retained_set
                == ORACLE_RETAINED_ALL
            )


            SIMULATION_RECORDS.append(
                {
                    "information_multiplier":
                        information_multiplier,

                    "replication":
                        replication + 1,

                    "method":
                        method_name,

                    "false_elimination":
                        float(
                            false_elimination
                        ),

                    "screening_power":
                        screening_power,

                    "retained_set_size":
                        len(
                            retained_set
                        ),

                    "oracle_agreement":
                        float(
                            oracle_agreement
                        ),

                    "ambiguity_rate":
                        ambiguity_value,
                }
            )


        if (
            replication + 1
        ) % 25 == 0:

            print(
                "Replication",
                replication + 1,
                "of",
                N_REPLICATIONS,
            )


SIMULATION_RESULTS = pd.DataFrame(
    SIMULATION_RECORDS
)


SCREENING_RUNTIME = (
    time.perf_counter()
    - SCREENING_START
)


# ============================================================
# 18. MONTE CARLO SUMMARY
# ============================================================

def mean_and_mcse(
    values: pd.Series,
) -> tuple[
    float,
    float,
]:

    array = np.asarray(
        values,
        dtype=float,
    )


    array = array[
        np.isfinite(
            array
        )
    ]


    if len(
        array
    ) == 0:

        return (
            np.nan,
            np.nan,
        )


    mean_value = float(
        np.mean(
            array
        )
    )


    if len(
        array
    ) <= 1:

        mcse = np.nan

    else:

        mcse = float(
            np.std(
                array,
                ddof=1,
            )
            / np.sqrt(
                len(
                    array
                )
            )
        )


    return (
        mean_value,
        mcse,
    )


SUMMARY_ROWS = []


for (
    information_multiplier,
    method_name
), group in (
    SIMULATION_RESULTS.groupby(
        [
            "information_multiplier",
            "method",
        ]
    )
):

    (
        false_mean,
        false_mcse,
    ) = mean_and_mcse(
        group[
            "false_elimination"
        ]
    )


    (
        power_mean,
        power_mcse,
    ) = mean_and_mcse(
        group[
            "screening_power"
        ]
    )


    (
        retained_mean,
        retained_mcse,
    ) = mean_and_mcse(
        group[
            "retained_set_size"
        ]
    )


    (
        agreement_mean,
        agreement_mcse,
    ) = mean_and_mcse(
        group[
            "oracle_agreement"
        ]
    )


    (
        ambiguity_mean,
        ambiguity_mcse,
    ) = mean_and_mcse(
        group[
            "ambiguity_rate"
        ]
    )


    SUMMARY_ROWS.append(
        {
            "information_multiplier":
                information_multiplier,

            "method":
                method_name,

            "false_elimination_probability":
                false_mean,

            "false_elimination_mcse":
                false_mcse,

            "screening_power":
                power_mean,

            "screening_power_mcse":
                power_mcse,

            "mean_retained_set_size":
                retained_mean,

            "retained_set_size_mcse":
                retained_mcse,

            "oracle_agreement_probability":
                agreement_mean,

            "oracle_agreement_mcse":
                agreement_mcse,

            "ambiguity_rate":
                ambiguity_mean,

            "ambiguity_rate_mcse":
                ambiguity_mcse,

            "replications":
                len(
                    group
                ),

            "max_t_draws_per_replication":
                N_BOOTSTRAP_DRAWS,
        }
    )


SCREENING_SUMMARY = pd.DataFrame(
    SUMMARY_ROWS
)


SCREENING_SUMMARY = (
    SCREENING_SUMMARY
    .sort_values(
        [
            "information_multiplier",
            "method",
        ]
    )
    .reset_index(
        drop=True
    )
)


print()
print(
    "Repeated-sampling intervention-screening results"
)

display(
    SCREENING_SUMMARY.round(
        4
    )
)


# ============================================================
# 19. FIGURE 1
#
# Policy curves and false-elimination performance
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        13,
        5,
    ),
)


# Policy risk curves

for policy_name, result in (
    FITTED_POLICIES.items()
):

    linewidth = (
        2.8
        if policy_name
        == BASELINE_POLICY
        else 1.8
    )


    axes[0].plot(
        THRESHOLDS,
        result[
            "risk_vector"
        ],
        marker="o",
        linewidth=
            linewidth,
        label=
            policy_name,
    )


axes[0].set_xlabel(
    "Final-size threshold"
)

axes[0].set_ylabel(
    r"$\Pr(N^I>c)$"
)

axes[0].set_ylim(
    -0.01,
    1.01,
)

axes[0].grid(
    alpha=0.20
)

axes[0].legend(
    frameon=False,
    fontsize=7,
)


# False elimination

for method_name, group in (
    SCREENING_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    axes[1].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "false_elimination_probability"
        ],
        yerr=
            1.96
            * group[
                "false_elimination_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=4,
        label=
            method_name,
    )


axes[1].axhline(
    ALPHA,
    color="black",
    linestyle="--",
    linewidth=1.2,
    label="0.05 reference",
)


axes[1].set_xlabel(
    "Relative information"
)

axes[1].set_ylabel(
    "False-elimination probability"
)

axes[1].set_ylim(
    -0.01,
    1.01,
)

axes[1].grid(
    alpha=0.20
)

axes[1].legend(
    frameon=False
)


fig.tight_layout()


fig.savefig(
    OUTPUT_DIR
    / "experiment3_policy_curves_false_elimination.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 20. FIGURE 2
#
# Screening power, retained size,
# oracle agreement, and ambiguity
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(
        12,
        9,
    ),
)


for method_name, group in (
    SCREENING_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    axes[
        0,
        0,
    ].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "screening_power"
        ],
        yerr=
            1.96
            * group[
                "screening_power_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method_name,
    )


    axes[
        0,
        1,
    ].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "mean_retained_set_size"
        ],
        yerr=
            1.96
            * group[
                "retained_set_size_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method_name,
    )


    axes[
        1,
        0,
    ].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "oracle_agreement_probability"
        ],
        yerr=
            1.96
            * group[
                "oracle_agreement_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method_name,
    )


    axes[
        1,
        1,
    ].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "ambiguity_rate"
        ],
        yerr=
            1.96
            * group[
                "ambiguity_rate_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method_name,
    )


axes[
    0,
    0,
].set_ylabel(
    "Screening power"
)

axes[
    0,
    0,
].set_ylim(
    -0.01,
    1.01,
)


axes[
    0,
    1,
].set_ylabel(
    "Mean retained-set size"
)


axes[
    1,
    0,
].set_ylabel(
    "Oracle-agreement probability"
)

axes[
    1,
    0,
].set_ylim(
    -0.01,
    1.01,
)


axes[
    1,
    1,
].set_ylabel(
    "Ambiguity rate"
)

axes[
    1,
    1,
].set_ylim(
    -0.01,
    1.01,
)


for axis in axes.ravel():

    axis.set_xlabel(
        "Relative information"
    )

    axis.grid(
        alpha=0.20
    )

    axis.legend(
        frameon=False,
        fontsize=8,
    )


fig.tight_layout()


fig.savefig(
    OUTPUT_DIR
    / "experiment3_screening_performance.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 21. SAVE RESULTS
# ============================================================

FIXED_INPUTS = pd.DataFrame(
    {
        "quantity": [

            "N_eff",
            "S0",
            "I0",
            "R0",

            "b_hat",
            "gamma_hat",
            "phi_hat",

            "exact_penalized_nll",

            "alpha",

            "Monte_Carlo_replications",

            "one_sided_max_t_draws",
        ],

        "value": [

            N,
            S0,
            I0,
            R0_INITIAL,

            B_HAT,
            GAMMA_HAT,
            PHI_HAT,

            EXACT_PENALIZED_NLL,

            ALPHA,

            N_REPLICATIONS,

            N_BOOTSTRAP_DRAWS,
        ],
    }
)


FIXED_INPUTS.to_csv(
    OUTPUT_DIR
    / "experiment3_fixed_inputs.csv",
    index=False,
)


APPLICATION_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_policy_descriptors.csv",
    index=False,
)


APPLICATION_GATE_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_application_gate_summary.csv",
    index=False,
)


PAIRWISE_GATE_RESULTS.to_csv(
    OUTPUT_DIR
    / "experiment3_pairwise_gate_results.csv",
    index=False,
)


BASELINE_CONTRAST_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_baseline_contrasts.csv",
    index=False,
)


SIMULATION_RESULTS.to_csv(
    OUTPUT_DIR
    / "experiment3_replication_level_results.csv",
    index=False,
)


SCREENING_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_screening_summary.csv",
    index=False,
)


np.savetxt(
    OUTPUT_DIR
    / "experiment3_fixed_Sigma_eta.csv",
    SIGMA_ETA,
    delimiter=",",
)


np.savetxt(
    OUTPUT_DIR
    / "experiment3_fixed_Hessian_eta.csv",
    HESSIAN_ETA,
    delimiter=",",
)


np.savetxt(
    OUTPUT_DIR
    / "experiment3_fixed_Sigma_log_b_log_gamma.csv",
    SIGMA_BG,
    delimiter=",",
)


# ============================================================
# 22. RUNTIME
# ============================================================

TOTAL_RUNTIME = (
    time.perf_counter()
    - TOTAL_START
)


RUNTIME_SUMMARY = pd.DataFrame(
    {
        "component": [

            "Repeated-sampling screening",

            "Total",
        ],

        "seconds": [

            SCREENING_RUNTIME,

            TOTAL_RUNTIME,
        ],
    }
)


RUNTIME_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_execution_times.csv",
    index=False,
)


print()
print(
    "Execution times"
)

display(
    RUNTIME_SUMMARY.round(
        3
    )
)


print()
print(
    "Experiment 3 completed"
)

print(
    "Results saved in",
    OUTPUT_DIR,
)
```

```python
# ============================================================
# EXPERIMENT 3
#
# LOCAL NEAR-BOUNDARY INTERVENTION SCREENING
#
# Purpose
#
# Compare
#
# 1. Plug-in dominance
# 2. Uncertainty-aware global one-sided max-t dominance
#
# The experiment is anchored at the fitted Cameroon model
# from Experiment 1.
#
# IMPORTANT
#
# Multiplicative interventions were replaced by smooth
# additive rate interventions.
#
# Under multiplicative interventions in the single-regime
# SIR model, final-size risk depends on b/gamma and policy
# ordering is essentially invariant to the common baseline
# parameter uncertainty.
#
# The additive intervention maps below create a genuine
# near-boundary decision problem while retaining the same
# fitted baseline model and final-size risk functional.
#
# No data loading
# No parameter fitting
# No Hessian estimation
#
# One Google Colab cell
# ============================================================


from __future__ import annotations

import time
from itertools import combinations
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from IPython.display import display


TOTAL_START = time.perf_counter()


# ============================================================
# 1. FIXED CAMEROON FIT FROM EXPERIMENT 1
# ============================================================

N = 175
S0 = 114
I0 = 1
R0_INITIAL = 60

assert N == S0 + I0 + R0_INITIAL


B_HAT = 0.7650254997116541
GAMMA_HAT = 0.23768509187018172
PHI_HAT = 2.4906108661339097


ETA_TRUE = np.log(
    [
        B_HAT,
        GAMMA_HAT,
    ]
)


SIGMA_ETA = np.array(
    [
        [
            0.01078625,
            0.01032878,
            -0.01802093,
        ],
        [
            0.01032878,
            0.01670211,
            -0.03299136,
        ],
        [
            -0.01802093,
            -0.03299136,
            0.55918411,
        ],
    ],
    dtype=float,
)


SIGMA_BG = SIGMA_ETA[
    np.ix_(
        [0, 1],
        [0, 1],
    )
]


# ============================================================
# 2. FINAL-SIZE THRESHOLDS
# ============================================================

THRESHOLD_FRACTIONS = np.array(
    [
        0.50,
        0.60,
        0.70,
        0.80,
        0.90,
    ],
    dtype=float,
)


THRESHOLDS = np.round(
    THRESHOLD_FRACTIONS
    * S0
).astype(int)


assert np.array_equal(
    THRESHOLDS,
    np.array(
        [
            57,
            68,
            80,
            91,
            103,
        ]
    ),
)


# Experiment 1 baseline risks used as a check

EXP1_TAIL_RISKS = np.array(
    [
        0.502185,
        0.492970,
        0.454210,
        0.318310,
        0.045407,
    ]
)


# ============================================================
# 3. MONTE CARLO SETTINGS
# ============================================================

ALPHA = 0.05

BASE_SEED = 20260807


# 500 is appropriate for manuscript results.

N_REPLICATIONS = 500


# Gaussian max-t draws within each replication.
#
# This is inexpensive because the dimension is only two.

N_MAXT_DRAWS = 999


# The experiment deliberately spans weak through strong
# information regimes.

INFORMATION_MULTIPLIERS = np.array(
    [
        0.25,
        0.50,
        1.00,
        2.00,
        4.00,
        8.00,
    ],
    dtype=float,
)


OUTPUT_DIR = Path(
    "/content/experiment3_near_boundary_screening"
)

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# ============================================================
# 4. LOCAL INTERVENTION DESIGN
#
# Transformation
#
# b_a     = b     - delta_b
# gamma_a = gamma + delta_gamma
#
# These are fixed absolute rate changes.
#
# They are calibrated once around the fitted Cameroon model.
# They are NOT changed after seeing the Monte Carlo results.
#
# Several policies are intentionally close in b_a/gamma_a.
# This creates the near-boundary comparison required to
# evaluate false elimination.
# ============================================================

POLICIES = {

    "No additional intervention": {
        "delta_b": 0.000,
        "delta_gamma": 0.000,
    },


    # Close candidate policies

    "Contact-focused response": {
        "delta_b": 0.100,
        "delta_gamma": 0.000,
    },

    "Recovery-focused response": {
        "delta_b": 0.000,
        "delta_gamma": 0.040,
    },

    "Balanced response A": {
        "delta_b": 0.050,
        "delta_gamma": 0.020,
    },

    "Balanced response B": {
        "delta_b": 0.070,
        "delta_gamma": 0.012,
    },


    # A more clearly dominated policy provides information
    # about screening power.

    "Weak response": {
        "delta_b": 0.030,
        "delta_gamma": 0.005,
    },
}


BASELINE_POLICY = (
    "No additional intervention"
)


CANDIDATE_POLICIES = [
    name
    for name in POLICIES
    if name != BASELINE_POLICY
]


# ============================================================
# 5. EXACT FINAL-SIZE RECURSION WITH SENSITIVITIES
#
# Local derivatives are with respect to
#
# log b_policy
# log gamma_policy
# ============================================================

def final_size_pmf_gradient(
    b: float,
    gamma: float,
):

    probability = np.zeros(
        I0 + 1,
        dtype=float,
    )

    probability[I0] = 1.0


    gradient = np.zeros(
        (
            2,
            I0 + 1,
        ),
        dtype=float,
    )


    pmf = np.zeros(
        S0 + 1,
        dtype=float,
    )


    pmf_gradient = np.zeros(
        (
            2,
            S0 + 1,
        ),
        dtype=float,
    )


    for susceptible in range(
        S0,
        -1,
        -1,
    ):

        maximum_infectious = (
            I0
            + S0
            - susceptible
        )


        required_length = (
            maximum_infectious
            + 1
        )


        if len(
            probability
        ) < required_length:

            padding = (
                required_length
                - len(
                    probability
                )
            )


            probability = np.pad(
                probability,
                (
                    0,
                    padding,
                ),
            )


            gradient = np.pad(
                gradient,
                (
                    (0, 0),
                    (0, padding),
                ),
            )


        if susceptible > 0:

            infection_component = (
                b
                * susceptible
                / N
            )


            denominator = (
                infection_component
                + gamma
            )


            infection_probability = (
                infection_component
                / denominator
            )


            recovery_probability = (
                gamma
                / denominator
            )


            derivative_infection = np.array(
                [
                    infection_probability
                    * recovery_probability,

                    -infection_probability
                    * recovery_probability,
                ],
                dtype=float,
            )


            derivative_recovery = (
                -derivative_infection
            )


            next_probability = np.zeros(
                maximum_infectious
                + 2,
                dtype=float,
            )


            next_gradient = np.zeros(
                (
                    2,
                    maximum_infectious
                    + 2,
                ),
                dtype=float,
            )


        else:

            infection_probability = 0.0
            recovery_probability = 1.0

            derivative_infection = np.zeros(
                2
            )

            derivative_recovery = np.zeros(
                2
            )

            next_probability = None
            next_gradient = None


        for infectious in range(
            maximum_infectious,
            0,
            -1,
        ):

            mass = probability[
                infectious
            ]


            mass_gradient = gradient[
                :,
                infectious,
            ].copy()


            if (
                abs(mass) < 1e-300
                and np.all(
                    np.abs(
                        mass_gradient
                    )
                    < 1e-300
                )
            ):

                continue


            # Recovery

            probability[
                infectious - 1
            ] += (
                mass
                * recovery_probability
            )


            gradient[
                :,
                infectious - 1,
            ] += (

                mass_gradient
                * recovery_probability

                + mass
                * derivative_recovery
            )


            # Infection

            if susceptible > 0:

                next_probability[
                    infectious + 1
                ] += (
                    mass
                    * infection_probability
                )


                next_gradient[
                    :,
                    infectious + 1,
                ] += (

                    mass_gradient
                    * infection_probability

                    + mass
                    * derivative_infection
                )


        infections_so_far = (
            S0
            - susceptible
        )


        pmf[
            infections_so_far
        ] = (
            probability[0]
        )


        pmf_gradient[
            :,
            infections_so_far,
        ] = gradient[
            :,
            0,
        ]


        if susceptible > 0:

            probability = (
                next_probability
            )

            gradient = (
                next_gradient
            )


    normalizer = float(
        pmf.sum()
    )


    normalizer_gradient = (
        pmf_gradient.sum(
            axis=1
        )
    )


    pmf_normalized = (
        pmf
        / normalizer
    )


    gradient_normalized = (

        pmf_gradient
        * normalizer

        - pmf[
            None,
            :
        ]
        * normalizer_gradient[
            :,
            None,
        ]

    ) / (
        normalizer**2
    )


    return (
        pmf_normalized,
        gradient_normalized,
    )


# ============================================================
# 6. STRICT TAIL RISK
# ============================================================

def strict_tail_and_gradient(
    pmf,
    pmf_gradient,
):

    nonstrict_tail = np.cumsum(
        pmf[::-1]
    )[::-1]


    nonstrict_gradient = np.cumsum(
        pmf_gradient[
            :,
            ::-1,
        ],
        axis=1,
    )[
        :,
        ::-1,
    ]


    strict_tail = (
        nonstrict_tail
        - pmf
    )


    strict_gradient = (
        nonstrict_gradient
        - pmf_gradient
    )


    return (
        strict_tail,
        strict_gradient,
    )


# ============================================================
# 7. POLICY TRANSFORMATION AND CHAIN RULE
# ============================================================

def evaluate_policy(
    eta_bg,
    policy_name,
):

    baseline_b = float(
        np.exp(
            eta_bg[0]
        )
    )


    baseline_gamma = float(
        np.exp(
            eta_bg[1]
        )
    )


    specification = (
        POLICIES[
            policy_name
        ]
    )


    delta_b = specification[
        "delta_b"
    ]

    delta_gamma = specification[
        "delta_gamma"
    ]


    policy_b = (
        baseline_b
        - delta_b
    )


    policy_gamma = (
        baseline_gamma
        + delta_gamma
    )


    if policy_b <= 0.0:

        raise RuntimeError(
            "An intervention produced a nonpositive "
            "transmission rate."
        )


    (
        pmf,
        pmf_gradient_local,
    ) = final_size_pmf_gradient(
        b=policy_b,
        gamma=policy_gamma,
    )


    (
        tail,
        tail_gradient_local,
    ) = strict_tail_and_gradient(
        pmf,
        pmf_gradient_local,
    )


    risk_vector = (
        tail[
            THRESHOLDS
        ]
    )


    local_gradient = (
        tail_gradient_local[
            :,
            THRESHOLDS,
        ].T
    )


    # --------------------------------------------------------
    # Chain rule
    #
    # policy_b = b - delta_b
    #
    # d log(policy_b) / d log(b)
    #
    # = b / policy_b
    #
    # policy_gamma = gamma + delta_gamma
    #
    # d log(policy_gamma) / d log(gamma)
    #
    # = gamma / policy_gamma
    # --------------------------------------------------------

    chain_scale = np.array(
        [
            baseline_b
            / policy_b,

            baseline_gamma
            / policy_gamma,
        ],
        dtype=float,
    )


    risk_gradient = (
        local_gradient
        * chain_scale[
            None,
            :
        ]
    )


    return {

        "policy":
            policy_name,

        "b":
            policy_b,

        "gamma":
            policy_gamma,

        "ratio":
            policy_b
            / policy_gamma,

        "risk_vector":
            risk_vector,

        "risk_gradient":
            risk_gradient,
    }


def evaluate_all_policies(
    eta_bg,
):

    return {

        policy_name:
            evaluate_policy(
                eta_bg,
                policy_name,
            )

        for policy_name
        in POLICIES
    }


# ============================================================
# 8. ORACLE POLICY CURVES
# ============================================================

TRUE_POLICY_RESULTS = (
    evaluate_all_policies(
        ETA_TRUE
    )
)


# Baseline check against Experiment 1

baseline_difference = np.max(
    np.abs(
        TRUE_POLICY_RESULTS[
            BASELINE_POLICY
        ][
            "risk_vector"
        ]
        - EXP1_TAIL_RISKS
    )
)


print()
print(
    "Maximum baseline difference from Experiment 1"
)

print(
    baseline_difference
)


if baseline_difference > 5e-4:

    raise RuntimeError(
        "The final-size recursion does not reproduce "
        "Experiment 1."
    )


# ============================================================
# 9. POLICY DESIGN TABLE
# ============================================================

DESIGN_ROWS = []


for policy_name, result in (
    TRUE_POLICY_RESULTS.items()
):

    row = {

        "policy":
            policy_name,

        "b":
            result[
                "b"
            ],

        "gamma":
            result[
                "gamma"
            ],

        "b_over_gamma":
            result[
                "ratio"
            ],
    }


    for index, threshold in enumerate(
        THRESHOLDS
    ):

        row[
            f"risk_{threshold}"
        ] = (
            result[
                "risk_vector"
            ][
                index
            ]
        )


    DESIGN_ROWS.append(
        row
    )


POLICY_DESIGN = pd.DataFrame(
    DESIGN_ROWS
)


print()
print(
    "Local intervention design"
)

display(
    POLICY_DESIGN.round(
        5
    )
)


# ============================================================
# 10. PLUG-IN DOMINANCE RULE
# ============================================================

def dominates_vector(
    first,
    second,
    tolerance=1e-10,
):

    return bool(

        np.all(
            first
            <= second
            + tolerance
        )

        and np.any(
            first
            < second
            - tolerance
        )
    )


def plugin_gate(
    policy_names,
    policy_results,
):

    eliminated = set()

    rows = []


    for first in policy_names:

        for second in policy_names:

            if first == second:

                continue


            contrast = (

                policy_results[
                    first
                ][
                    "risk_vector"
                ]

                - policy_results[
                    second
                ][
                    "risk_vector"
                ]
            )


            dominance = (
                dominates_vector(

                    policy_results[
                        first
                    ][
                        "risk_vector"
                    ],

                    policy_results[
                        second
                    ][
                        "risk_vector"
                    ],
                )
            )


            if dominance:

                eliminated.add(
                    second
                )


            rows.append(
                {
                    "potential_dominator":
                        first,

                    "potentially_dominated":
                        second,

                    "maximum_contrast":
                        float(
                            np.max(
                                contrast
                            )
                        ),

                    "plugin_dominance":
                        dominance,
                }
            )


    retained = (
        set(
            policy_names
        )
        - eliminated
    )


    return (
        retained,
        pd.DataFrame(
            rows
        ),
    )


# ============================================================
# 11. GLOBAL ONE-SIDED MAX-T ROBUST GATE
# ============================================================

def robust_gate(
    policy_names,
    policy_results,
    covariance,
    rng,
    n_draws,
):

    parameter_draws = (
        rng.multivariate_normal(
            mean=np.zeros(
                2
            ),
            cov=covariance,
            size=n_draws,
        )
    )


    pair_information = []

    standardized_blocks = []


    for first in policy_names:

        for second in policy_names:

            if first == second:

                continue


            contrast = (

                policy_results[
                    first
                ][
                    "risk_vector"
                ]

                - policy_results[
                    second
                ][
                    "risk_vector"
                ]
            )


            contrast_gradient = (

                policy_results[
                    first
                ][
                    "risk_gradient"
                ]

                - policy_results[
                    second
                ][
                    "risk_gradient"
                ]
            )


            variance = np.maximum(
                np.einsum(
                    "ki,ij,kj->k",
                    contrast_gradient,
                    covariance,
                    contrast_gradient,
                ),
                0.0,
            )


            standard_error = np.sqrt(
                variance
            )


            active = (
                standard_error
                > 1e-12
            )


            perturbation = (

                parameter_draws

                @ contrast_gradient.T
            )


            if np.any(
                active
            ):

                standardized_blocks.append(

                    perturbation[
                        :,
                        active,
                    ]

                    / standard_error[
                        active
                    ]
                )


            pair_information.append(
                {
                    "first":
                        first,

                    "second":
                        second,

                    "contrast":
                        contrast,

                    "standard_error":
                        standard_error,
                }
            )


    if standardized_blocks:

        standardized_all = np.concatenate(
            standardized_blocks,
            axis=1,
        )


        max_statistic = np.max(
            standardized_all,
            axis=1,
        )


        critical_value = float(
            np.quantile(
                max_statistic,
                1.0 - ALPHA,
            )
        )


    else:

        critical_value = 0.0


    eliminated = set()

    rows = []


    for information in (
        pair_information
    ):

        contrast = information[
            "contrast"
        ]


        standard_error = information[
            "standard_error"
        ]


        upper_band = (

            contrast

            + critical_value
            * standard_error
        )


        dominance = bool(

            np.all(
                upper_band
                <= 1e-10
            )

            and np.any(
                upper_band
                < -1e-10
            )
        )


        if dominance:

            eliminated.add(
                information[
                    "second"
                ]
            )


        rows.append(
            {
                "potential_dominator":
                    information[
                        "first"
                    ],

                "potentially_dominated":
                    information[
                        "second"
                    ],

                "maximum_contrast":
                    float(
                        np.max(
                            contrast
                        )
                    ),

                "maximum_upper_band":
                    float(
                        np.max(
                            upper_band
                        )
                    ),

                "robust_dominance":
                    dominance,
            }
        )


    retained = (
        set(
            policy_names
        )
        - eliminated
    )


    return (
        retained,
        pd.DataFrame(
            rows
        ),
        critical_value,
    )


# ============================================================
# 12. AMBIGUITY
# ============================================================

def ambiguity_rate(
    policy_names,
    pairwise_table,
    dominance_column,
):

    values = []


    for first, second in combinations(
        policy_names,
        2,
    ):

        first_dominates = bool(

            pairwise_table.loc[
                (
                    pairwise_table[
                        "potential_dominator"
                    ] == first
                )
                & (
                    pairwise_table[
                        "potentially_dominated"
                    ] == second
                ),
                dominance_column,
            ].iloc[0]
        )


        second_dominates = bool(

            pairwise_table.loc[
                (
                    pairwise_table[
                        "potential_dominator"
                    ] == second
                )
                & (
                    pairwise_table[
                        "potentially_dominated"
                    ] == first
                ),
                dominance_column,
            ].iloc[0]
        )


        values.append(

            not (
                first_dominates
                or second_dominates
            )
        )


    return float(
        np.mean(
            values
        )
    )


# ============================================================
# 13. ORACLE DOMINANCE STRUCTURE
# ============================================================

(
    ORACLE_RETAINED,
    ORACLE_PAIRWISE,
) = plugin_gate(
    CANDIDATE_POLICIES,
    TRUE_POLICY_RESULTS,
)


ORACLE_DOMINATED = (

    set(
        CANDIDATE_POLICIES
    )

    - ORACLE_RETAINED
)


print()
print(
    "Oracle retained intervention set"
)

print(
    sorted(
        ORACLE_RETAINED
    )
)


print()
print(
    "Oracle dominated interventions"
)

print(
    sorted(
        ORACLE_DOMINATED
    )
)


# ============================================================
# 14. FITTED CAMEROON GATE
# ============================================================

APPLICATION_RNG = np.random.default_rng(
    BASE_SEED
)


(
    APPLICATION_PLUGIN_RETAINED,
    APPLICATION_PLUGIN_TABLE,
) = plugin_gate(
    CANDIDATE_POLICIES,
    TRUE_POLICY_RESULTS,
)


(
    APPLICATION_ROBUST_RETAINED,
    APPLICATION_ROBUST_TABLE,
    APPLICATION_CRITICAL_VALUE,
) = robust_gate(
    CANDIDATE_POLICIES,
    TRUE_POLICY_RESULTS,
    SIGMA_BG,
    APPLICATION_RNG,
    10000,
)


print()
print(
    "Plug-in retained set at fitted Cameroon parameters"
)

print(
    sorted(
        APPLICATION_PLUGIN_RETAINED
    )
)


print()
print(
    "Robust retained set at fitted Cameroon parameters"
)

print(
    sorted(
        APPLICATION_ROBUST_RETAINED
    )
)


print()
print(
    "Global one-sided max-t critical value"
)

print(
    APPLICATION_CRITICAL_VALUE
)


# ============================================================
# 15. REPEATED-SAMPLING STRESS TEST
#
# eta_hat is generated around the fitted Cameroon parameter.
#
# The intervention definitions remain fixed.
#
# The only quantity changed across information regimes is
# estimator covariance.
# ============================================================

SIMULATION_START = time.perf_counter()


SIMULATION_RNG = np.random.default_rng(
    BASE_SEED
    + 10000
)


SIMULATION_ROWS = []


for information_multiplier in (
    INFORMATION_MULTIPLIERS
):

    covariance = (

        SIGMA_BG

        / information_multiplier
    )


    print()
    print(
        "Relative information",
        information_multiplier,
    )


    for replication in range(
        N_REPLICATIONS
    ):

        eta_estimate = (
            SIMULATION_RNG.multivariate_normal(
                mean=
                    ETA_TRUE,
                cov=
                    covariance,
            )
        )


        estimated_results = (
            evaluate_all_policies(
                eta_estimate
            )
        )


        # ----------------------------------------------------
        # Plug-in
        # ----------------------------------------------------

        (
            plugin_retained,
            plugin_table,
        ) = plugin_gate(
            CANDIDATE_POLICIES,
            estimated_results,
        )


        plugin_eliminated = (

            set(
                CANDIDATE_POLICIES
            )

            - plugin_retained
        )


        plugin_ambiguity = (
            ambiguity_rate(
                CANDIDATE_POLICIES,
                plugin_table,
                "plugin_dominance",
            )
        )


        # ----------------------------------------------------
        # Robust
        # ----------------------------------------------------

        (
            robust_retained,
            robust_table,
            robust_critical_value,
        ) = robust_gate(
            CANDIDATE_POLICIES,
            estimated_results,
            covariance,
            SIMULATION_RNG,
            N_MAXT_DRAWS,
        )


        robust_eliminated = (

            set(
                CANDIDATE_POLICIES
            )

            - robust_retained
        )


        robust_ambiguity = (
            ambiguity_rate(
                CANDIDATE_POLICIES,
                robust_table,
                "robust_dominance",
            )
        )


        # ----------------------------------------------------
        # Metrics
        # ----------------------------------------------------

        for (
            method,
            retained,
            eliminated,
            ambiguity,
        ) in [

            (
                "Plug-in gate",
                plugin_retained,
                plugin_eliminated,
                plugin_ambiguity,
            ),

            (
                "Robust simultaneous gate",
                robust_retained,
                robust_eliminated,
                robust_ambiguity,
            ),
        ]:


            false_elimination = (

                len(
                    eliminated
                    & ORACLE_RETAINED
                )

                > 0
            )


            screening_power = (

                len(
                    eliminated
                    & ORACLE_DOMINATED
                )

                / len(
                    ORACLE_DOMINATED
                )
            )


            exact_oracle_set = (

                retained
                == ORACLE_RETAINED
            )


            SIMULATION_ROWS.append(
                {
                    "information_multiplier":
                        information_multiplier,

                    "replication":
                        replication + 1,

                    "method":
                        method,

                    "false_elimination":
                        float(
                            false_elimination
                        ),

                    "screening_power":
                        screening_power,

                    "retained_set_size":
                        len(
                            retained
                        ),

                    "oracle_set_recovery":
                        float(
                            exact_oracle_set
                        ),

                    "ambiguity_rate":
                        ambiguity,

                    "critical_value":
                        (
                            robust_critical_value
                            if method
                            == "Robust simultaneous gate"
                            else np.nan
                        ),
                }
            )


        if (
            replication + 1
        ) % 100 == 0:

            print(
                "Replication",
                replication + 1,
                "of",
                N_REPLICATIONS,
            )


SIMULATION_RESULTS = pd.DataFrame(
    SIMULATION_ROWS
)


SIMULATION_RUNTIME = (
    time.perf_counter()
    - SIMULATION_START
)


# ============================================================
# 16. MONTE CARLO SUMMARIES
# ============================================================

def summarize_metric(
    values,
):

    values = np.asarray(
        values,
        dtype=float,
    )


    values = values[
        np.isfinite(
            values
        )
    ]


    mean = float(
        np.mean(
            values
        )
    )


    mcse = float(

        np.std(
            values,
            ddof=1,
        )

        / np.sqrt(
            len(
                values
            )
        )
    )


    return (
        mean,
        mcse,
    )


SUMMARY_ROWS = []


for (
    information_multiplier,
    method
), group in (
    SIMULATION_RESULTS.groupby(
        [
            "information_multiplier",
            "method",
        ]
    )
):


    false_mean, false_mcse = (
        summarize_metric(
            group[
                "false_elimination"
            ]
        )
    )


    power_mean, power_mcse = (
        summarize_metric(
            group[
                "screening_power"
            ]
        )
    )


    retained_mean, retained_mcse = (
        summarize_metric(
            group[
                "retained_set_size"
            ]
        )
    )


    oracle_mean, oracle_mcse = (
        summarize_metric(
            group[
                "oracle_set_recovery"
            ]
        )
    )


    ambiguity_mean, ambiguity_mcse = (
        summarize_metric(
            group[
                "ambiguity_rate"
            ]
        )
    )


    SUMMARY_ROWS.append(
        {
            "information_multiplier":
                information_multiplier,

            "method":
                method,

            "false_elimination_probability":
                false_mean,

            "false_elimination_mcse":
                false_mcse,

            "screening_power":
                power_mean,

            "screening_power_mcse":
                power_mcse,

            "mean_retained_set_size":
                retained_mean,

            "retained_set_size_mcse":
                retained_mcse,

            "oracle_set_recovery":
                oracle_mean,

            "oracle_set_recovery_mcse":
                oracle_mcse,

            "ambiguity_rate":
                ambiguity_mean,

            "ambiguity_rate_mcse":
                ambiguity_mcse,
        }
    )


SCREENING_SUMMARY = pd.DataFrame(
    SUMMARY_ROWS
)


SCREENING_SUMMARY = (
    SCREENING_SUMMARY
    .sort_values(
        [
            "information_multiplier",
            "method",
        ]
    )
    .reset_index(
        drop=True
    )
)


print()
print(
    "Repeated-sampling screening results"
)

display(
    SCREENING_SUMMARY.round(
        4
    )
)


# ============================================================
# 17. DIRECT METHOD COMPARISON
# ============================================================

plugin_summary = (
    SCREENING_SUMMARY.loc[
        SCREENING_SUMMARY[
            "method"
        ]
        == "Plug-in gate"
    ]
    .set_index(
        "information_multiplier"
    )
)


robust_summary = (
    SCREENING_SUMMARY.loc[
        SCREENING_SUMMARY[
            "method"
        ]
        == "Robust simultaneous gate"
    ]
    .set_index(
        "information_multiplier"
    )
)


COMPARISON_SUMMARY = pd.DataFrame(
    {
        "information_multiplier":
            INFORMATION_MULTIPLIERS,

        "plugin_false_elimination":
            plugin_summary.loc[
                INFORMATION_MULTIPLIERS,
                "false_elimination_probability",
            ].to_numpy(),

        "robust_false_elimination":
            robust_summary.loc[
                INFORMATION_MULTIPLIERS,
                "false_elimination_probability",
            ].to_numpy(),

        "false_elimination_reduction_percentage_points":
            100.0
            * (
                plugin_summary.loc[
                    INFORMATION_MULTIPLIERS,
                    "false_elimination_probability",
                ].to_numpy()

                - robust_summary.loc[
                    INFORMATION_MULTIPLIERS,
                    "false_elimination_probability",
                ].to_numpy()
            ),

        "plugin_screening_power":
            plugin_summary.loc[
                INFORMATION_MULTIPLIERS,
                "screening_power",
            ].to_numpy(),

        "robust_screening_power":
            robust_summary.loc[
                INFORMATION_MULTIPLIERS,
                "screening_power",
            ].to_numpy(),

        "plugin_oracle_recovery":
            plugin_summary.loc[
                INFORMATION_MULTIPLIERS,
                "oracle_set_recovery",
            ].to_numpy(),

        "robust_oracle_recovery":
            robust_summary.loc[
                INFORMATION_MULTIPLIERS,
                "oracle_set_recovery",
            ].to_numpy(),
    }
)


print()
print(
    "Direct plug-in versus robust comparison"
)

display(
    COMPARISON_SUMMARY.round(
        4
    )
)


# ============================================================
# 18. FIGURE 1
#
# Local risk curves and false elimination
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(
        13,
        5,
    ),
)


# ------------------------------------------------------------
# Local intervention curves
# ------------------------------------------------------------

for policy_name in (
    POLICIES
):

    result = (
        TRUE_POLICY_RESULTS[
            policy_name
        ]
    )


    linewidth = (
        2.8
        if policy_name
        == BASELINE_POLICY
        else 1.8
    )


    axes[0].plot(
        THRESHOLDS,
        result[
            "risk_vector"
        ],
        marker="o",
        linewidth=
            linewidth,
        label=
            policy_name,
    )


axes[0].set_xlabel(
    "Final-size threshold"
)

axes[0].set_ylabel(
    r"$\Pr(N^I>c)$"
)

axes[0].set_ylim(
    -0.01,
    0.60,
)

axes[0].grid(
    alpha=0.20
)

axes[0].legend(
    frameon=False,
    fontsize=7,
)


# ------------------------------------------------------------
# False elimination
# ------------------------------------------------------------

for method, group in (
    SCREENING_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    axes[1].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "false_elimination_probability"
        ],
        yerr=
            1.96
            * group[
                "false_elimination_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=4,
        label=
            method,
    )


axes[1].axhline(
    ALPHA,
    color="black",
    linestyle="--",
    linewidth=1.2,
    label="0.05 reference",
)


axes[1].set_xlabel(
    "Relative information"
)

axes[1].set_ylabel(
    "False-elimination probability"
)

axes[1].set_ylim(
    -0.01,
    0.65,
)

axes[1].grid(
    alpha=0.20
)

axes[1].legend(
    frameon=False
)


fig.tight_layout()


fig.savefig(
    OUTPUT_DIR
    / "experiment3_false_elimination.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 19. FIGURE 2
#
# Protection versus screening power
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(
        15,
        4.6,
    ),
)


for method, group in (
    SCREENING_SUMMARY.groupby(
        "method"
    )
):

    group = group.sort_values(
        "information_multiplier"
    )


    axes[0].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "screening_power"
        ],
        yerr=
            1.96
            * group[
                "screening_power_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method,
    )


    axes[1].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "mean_retained_set_size"
        ],
        yerr=
            1.96
            * group[
                "retained_set_size_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method,
    )


    axes[2].errorbar(
        group[
            "information_multiplier"
        ],
        group[
            "oracle_set_recovery"
        ],
        yerr=
            1.96
            * group[
                "oracle_set_recovery_mcse"
            ],
        marker="o",
        linewidth=2,
        capsize=3,
        label=
            method,
    )


axes[0].set_ylabel(
    "Screening power"
)

axes[1].set_ylabel(
    "Mean retained-set size"
)

axes[2].set_ylabel(
    "Oracle-set recovery probability"
)


for axis in axes:

    axis.set_xlabel(
        "Relative information"
    )

    axis.grid(
        alpha=0.20
    )

    axis.legend(
        frameon=False,
        fontsize=8,
    )


axes[0].set_ylim(
    -0.01,
    1.01,
)

axes[2].set_ylim(
    -0.01,
    1.01,
)


fig.tight_layout()


fig.savefig(
    OUTPUT_DIR
    / "experiment3_power_and_stability.png",
    dpi=300,
    bbox_inches="tight",
)


plt.show()


# ============================================================
# 20. SAVE RESULTS
# ============================================================

POLICY_DESIGN.to_csv(
    OUTPUT_DIR
    / "experiment3_policy_design.csv",
    index=False,
)


ORACLE_PAIRWISE.to_csv(
    OUTPUT_DIR
    / "experiment3_oracle_pairwise.csv",
    index=False,
)


APPLICATION_PLUGIN_TABLE.to_csv(
    OUTPUT_DIR
    / "experiment3_application_plugin_gate.csv",
    index=False,
)


APPLICATION_ROBUST_TABLE.to_csv(
    OUTPUT_DIR
    / "experiment3_application_robust_gate.csv",
    index=False,
)


SIMULATION_RESULTS.to_csv(
    OUTPUT_DIR
    / "experiment3_replication_results.csv",
    index=False,
)


SCREENING_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_screening_summary.csv",
    index=False,
)


COMPARISON_SUMMARY.to_csv(
    OUTPUT_DIR
    / "experiment3_method_comparison.csv",
    index=False,
)


np.savetxt(
    OUTPUT_DIR
    / "experiment3_Sigma_log_b_log_gamma.csv",
    SIGMA_BG,
    delimiter=",",
)


# ============================================================
# 21. FINAL DIAGNOSTICS
# ============================================================

TOTAL_RUNTIME = (
    time.perf_counter()
    - TOTAL_START
)


print()
print(
    "Experiment 3 completed"
)

print(
    {
        "oracle_retained":
            sorted(
                ORACLE_RETAINED
            ),

        "number_oracle_dominated":
            len(
                ORACLE_DOMINATED
            ),

        "application_plugin_retained":
            sorted(
                APPLICATION_PLUGIN_RETAINED
            ),

        "application_robust_retained":
            sorted(
                APPLICATION_ROBUST_RETAINED
            ),

        "runtime_seconds":
            TOTAL_RUNTIME,
    }
)


print()
print(
    "Results saved in",
    OUTPUT_DIR,
)
```
